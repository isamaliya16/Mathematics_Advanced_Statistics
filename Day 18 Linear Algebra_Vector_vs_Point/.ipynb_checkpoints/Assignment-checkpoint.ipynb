# 📘 Linear Algebra – 

## 1️⃣ What is Linear Algebra?

Linear Algebra is a branch of mathematics that deals with:
- Vectors
- Matrices
- Linear equations
- Vector spaces

It is the mathematical foundation of Machine Learning, Deep Learning, and Data Science.

---

## 2️⃣ Difference Between Point and Vector

| Point | Vector |
|-------|--------|
| Represents a location in space | Represents direction and magnitude |
| Example: (3, 4) | Example: ⟨3, 4⟩ |
| Fixed position | Has direction and length |

👉 A point tells *where something is*.  
👉 A vector tells *how to move*.

---

## 3️⃣ What is Magnitude of a Vector?

The magnitude (or length) of a vector measures how long the vector is.

For vector **v = (x, y)**:
||v|| = √(x² + y²)

Example:

For v = (3, 4)
||v|| = √(3² + 4²) = √(9 + 16) = 5

---

## 4️⃣ What Does Dot Product Measure?

The dot product measures:

- Similarity between two vectors
- How much one vector aligns with another

Formula:
A · B = |A||B| cosθ

If:
- Dot product > 0 → Vectors point in similar direction
- Dot product = 0 → Vectors are perpendicular
- Dot product < 0 → Vectors point in opposite direction

---

## 5️⃣ How Do We Compute Angle Between Two Vectors?

Using dot product formula:  \
cosθ = (A · B) / (|A||B|) \
θ = cos⁻¹((A · B) / (|A||B|))

This gives the angle between vectors.

---

## 6️⃣ Why Are Vectors Important in Machine Learning?

Vectors are important because:

- Data is represented as vectors
- Features are vectors
- Images are matrices (collection of vectors)
- Text is converted into vector embeddings
- Algorithms like Linear Regression, SVM, Neural Networks use vector operations

👉 Machine Learning = Linear Algebra + Statistics + Programming

# 📌 Mini Assignment

### A company tracks:
- Experience (years)
- Performance score

## Employees:

E1 = [5, 80]

E2 = [7, 85]

E3 = [3, 70]

### Tasks:
1. Represent employees as vectors.
2. Compute magnitude of each vector.
3. Compute dot product between E1 and E2.
4. Calculate angle between E1 and E2.
5. Identify most similar employees using cosine similarity.
6. Visualize vectors.
7. Interpret business meaning.


```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# Employee Data (Experience, Performance Score)
# ==============================

E1 = np.array([5, 80])
E2 = np.array([7, 85])
E3 = np.array([3, 70])

print("Employee Vectors:")
print("E1 =", E1)
print("E2 =", E2)
print("E3 =", E3)

# ==============================
# Magnitude of Vectors
# ==============================

mag_E1 = np.linalg.norm(E1)
mag_E2 = np.linalg.norm(E2)
mag_E3 = np.linalg.norm(E3)

print("\nMagnitudes:")
print("||E1|| =", round(mag_E1, 4))
print("||E2|| =", round(mag_E2, 4))
print("||E3|| =", round(mag_E3, 4))

# ==============================
# Dot Product
# ==============================

dot_product = np.dot(E1, E2)
print("\nDot Product (E1 · E2) =", dot_product)

# ==============================
# Angle Between Vectors
# ==============================

cos_theta = dot_product / (mag_E1 * mag_E2)
angle = np.degrees(np.arccos(cos_theta))

print("Angle between E1 and E2 (degrees) =", round(angle, 4))

# ==============================
# Cosine Similarity
# ==============================

cosine_similarity = cos_theta
print("Cosine Similarity (E1, E2) =", round(cosine_similarity, 6))

# ==============================
# Plotting Vectors
# ==============================

plt.figure()
plt.quiver(0, 0, E1[0], E1[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, E2[0], E2[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, E3[0], E3[1], angles='xy', scale_units='xy', scale=1)

plt.xlim(0, 10)
plt.ylim(0, 100)

plt.xlabel("Experience (Years)")
plt.ylabel("Performance Score")
plt.title("Employee Vectors Representation")
plt.grid()

plt.show()
```
