# 📊 Normal (Gaussian) Distribution – Complete Overview

---

# 2️. Normal (Gaussian) Distribution – Overview

## 2.1 Definition

The **Normal (Gaussian) Distribution** is a continuous probability distribution that is symmetric about its mean and forms a characteristic **bell-shaped curve**.

It describes how values of a variable are distributed around the mean, where:

- Most observations cluster near the center
- Fewer observations occur as we move away from the mean
- The distribution is completely determined by its mean and standard deviation

---

## 2.2 Historical Background

The Normal Distribution is named after **Carl Friedrich Gauss**, who introduced it while studying measurement errors.

It plays a central role in:

- Statistical inference  
- Hypothesis testing  
- Machine Learning models  
- Central Limit Theorem  
- Real-world data modeling  

Because of its importance, it is also known as the **Gaussian Distribution**.

---

## 2.3 Graphical Shape

The graph of a normal distribution has the following characteristics:

- 📈 Bell-shaped curve  
- 📍 Perfectly symmetrical  
- 📍 Mean = Median = Mode  
- 📍 Highest point at the mean  
- 📍 Tails extend infinitely but never touch the x-axis  

---

# 📐 3️. Mathematical Representation

![Probability_Density_Function](images/formula.png)

### Interpretation

- The curve is centered at μ  
- Spread is controlled by σ  
- Total area under the curve = 1  

If σ is small → Curve is narrow and tall  
If σ is large → Curve is wide and flat  

---

## 3.2 Parameters of Normal Distribution

A Normal Distribution is completely defined by two parameters:

### 1️. Mean (μ)
- Determines the center of the distribution
- Shifts the curve left or right

### 2️. Variance (σ²)
- Measures how spread out the data is

### 3️. Standard Deviation (σ)
- Square root of variance
- Controls the width of the bell curve

---

# 📌 4️. Properties of Normal Distribution

The Normal Distribution has the following properties:

- ✔ Symmetrical about the mean  
- ✔ Mean = Median = Mode  
- ✔ Total area under the curve = 1  
- ✔ 50% of data lies on each side of the mean  
- ✔ Tails approach the x-axis but never touch it  
- ✔ Defined for (-∞, +∞)  
- ✔ Completely determined by μ and σ  

---

# 🎯 5️. Standard Normal Distribution

![Standard_Normal_Distribution](images/Standard_Normal.png)

### Interpretation

- Z > 0 → Value is above the mean  
- Z < 0 → Value is below the mean  
- Z = 0 → Value equals the mean  

The Z-score tells us how many standard deviations a value is away from the mean.

---

## 5.3 Z-Table

The **Z-table** provides probabilities associated with Z-scores.

It helps calculate:

- 📍 Left-tail probability → P(Z < z)  
- 📍 Right-tail probability → P(Z > z)  
- 📍 Two-tailed probability → P(|Z| > z)  

Z-tables are widely used in:

- Hypothesis testing  
- Confidence intervals  
- Statistical decision-making  

---

# 📊 6️. Empirical Rule (68–95–99.7 Rule)

The Empirical Rule describes the percentage of data within standard deviation ranges.

| Range        | Percentage of Data |
|-------------|-------------------|
| μ ± 1σ      | 68%               |
| μ ± 2σ      | 95%               |
| μ ± 3σ      | 99.7%             |

### Explanation

- 68% of data lies within 1 standard deviation  
- 95% lies within 2 standard deviations  
- 99.7% lies within 3 standard deviations  

This rule is useful for:

- Outlier detection  
- Risk assessment  
- Quality control  

---

# 🔎 7️⃣ Types of Normal Distribution

## 7.1 Standard Normal Distribution
- Mean = 0  
- Standard Deviation = 1  
- Used in Z-score calculations  

---

## 7.2 Non-Standard Normal Distribution
- Mean ≠ 0  
- Standard Deviation ≠ 1  
- Most real-world datasets follow this type  

---

## 7.3 Multivariate Normal Distribution

Extension of normal distribution to multiple variables.

Defined using:
- Mean vector  
- Covariance matrix  

Used in:
- Machine Learning  
- Gaussian Mixture Models  
- Bayesian statistics  

---

## 7.4 Truncated Normal Distribution

A normal distribution limited within a specific range.

Used when:
- Data has natural bounds  
- Physical constraints exist  

---

## 7.5 Log-Normal Distribution 

A variable is log-normally distributed if its logarithm follows a normal distribution.

Common applications:

- Stock prices  
- Income distribution  
- Biological growth measurements  

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
normal_data = np.random.normal(loc=70, scale=10, size=1000)
df = pd.Series(normal_data)

mean = np.mean(normal_data).round(4)
median = np.median(normal_data).round(4)
mode = df.mode()[0].round(4)
std = np.std(normal_data).round(4)

print("mean : ",mean)
print("median : ",median)
print("mode : ",mode)
print("Standard Deviation : ",std)
```

```python

plt.figure()
plt.hist(normal_data, bins=30)
plt.title("Gaussian Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True, linestyle='--',alpha=0.6)
plt.tight_layout()
plt.show()
```


## 2️⃣ Empirical Rule (68–95–99.7 Rule)

- 68% within ±1σ  
- 95% within ±2σ  
- 99.7% within ±3σ  


```python

mean = np.mean(normal_data)
std = np.std(normal_data)

within_1 = np.mean((normal_data >= mean-std) & (normal_data <= mean+std))*100
within_2 = np.mean((normal_data >= mean-2*std) & (normal_data <= mean+2*std))*100
within_3 = np.mean((normal_data >= mean-3*std) & (normal_data <= mean+3*std))*100

print(f"Within 1σ (Mean ± 1 Std Dev): {within_1:.2f}%")
print(f"Within 2σ (Mean ± 2 Std Dev): {within_2:.2f}%")
print(f"Within 3σ (Mean ± 3 Std Dev): {within_3:.2f}%")
```


## 3️⃣ Skewness

Skewness measures **asymmetry**.

- Positive → Right tail longer  
- Negative → Left tail longer  


```python

from scipy.stats import skew

positive_skew = np.random.exponential(scale=10, size=1000)
negative_skew = -positive_skew + 40

print("Positive Skewness : ",skew(positive_skew).round(4))
print("Negative Skewness : ",skew(negative_skew).round(4))
```

```python


plt.figure(figsize=(14, 6))

# ---------------- Positive Skewness ----------------
plt.subplot(1, 2, 1)
plt.hist(positive_skew, bins=30, density=True, alpha=0.7)
plt.axvline(np.mean(positive_skew), linestyle='--', linewidth=2)
plt.title("Positive Skewness", fontsize=13, fontweight="bold")
plt.xlabel("Values")
plt.ylabel("Density")
plt.grid(True, linestyle="--", alpha=0.6)

# ---------------- Negative Skewness ----------------
plt.subplot(1, 2, 2)
plt.hist(negative_skew, bins=30, density=True, alpha=0.7)
plt.axvline(np.mean(negative_skew), linestyle='--', linewidth=2)
plt.title("Negative Skewness", fontsize=13, fontweight="bold")
plt.xlabel("Values")
plt.ylabel("Density")
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```


## 4️⃣ Kurtosis

Kurtosis measures **tailedness**.

- Mesokurtic → Normal  
- Leptokurtic → Heavy tails  
- Platykurtic → Flat  


```python
from scipy.stats import kurtosis

np.random.seed(42)

# Mesokurtic (Normal Distribution)
normal_data = np.random.normal(loc=0, scale=1, size=10000)

# Leptokurtic (Heavy tails) – Student's t distribution
heavy_tail = np.random.standard_t(df=2, size=10000)

# Platykurtic (Flat distribution) – Uniform distribution
flat_data = np.random.uniform(low=-4, high=4, size=10000)


print("Mesokurtic Kurtosis:", kurtosis(normal_data))
print("Leptokurtic Kurtosis:", kurtosis(heavy_tail))
print("Platykurtic Kurtosis:", kurtosis(flat_data))

```

```python
plt.figure(figsize=(18, 5))

# ---------------- Mesokurtic ----------------
plt.subplot(1, 3, 1)
plt.hist(normal_data, bins=30, density=True, alpha=0.7)
plt.axvline(np.mean(normal_data), linestyle='--', linewidth=2)
plt.title("Mesokurtic (Normal)", fontweight="bold")
plt.xlabel("Values")
plt.ylabel("Density")
plt.grid(True, linestyle="--", alpha=0.6)

# ---------------- Leptokurtic ----------------
plt.subplot(1, 3, 2)
plt.hist(heavy_tail, bins=30, density=True, alpha=0.7)
plt.axvline(np.mean(heavy_tail), linestyle='--', linewidth=2)
plt.title("Leptokurtic (Heavy Tails)", fontweight="bold")
plt.xlabel("Values")
plt.ylabel("Density")
plt.grid(True, linestyle="--", alpha=0.6)

# ---------------- Platykurtic ----------------
plt.subplot(1, 3, 3)
plt.hist(flat_data, bins=30, density=True, alpha=0.7)
plt.axvline(np.mean(flat_data), linestyle='--', linewidth=2)
plt.title("Platykurtic (Flat)", fontweight="bold")
plt.xlabel("Values")
plt.ylabel("Density")
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
