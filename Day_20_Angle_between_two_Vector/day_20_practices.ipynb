# 🔹 Part 1: Angle Between Two Vectors

Formula:
cos(θ) = (A · B) / (|A| |B|)

Real-World Example:
Customer feature vectors: [Age, Income]


```python

import numpy as np
import matplotlib.pyplot as plt

A = np.array([25, 45])
B = np.array([40, 60])

dot_product = np.dot(A, B)
mag_A = np.linalg.norm(A)
mag_B = np.linalg.norm(B)

cos_theta = dot_product / (mag_A * mag_B)
angle = np.degrees(np.arccos(cos_theta))

angle
```

```python

plt.quiver(0, 0, A[0], A[1], angles="xy", scale_units="xy", scale=1, color="blue")
plt.quiver(0, 0, B[0], B[1], angles="xy", scale_units="xy", scale=1, color="orange")
plt.xlim(0, 50)
plt.ylim(0, 70)
plt.title("Angle Between Two Vectors")
plt.show()
```


# 🔹 Part 2: Vector Projection

Vector projection tells us:

How much of one vector lies in the direction of another vector.

It answers:
	•	“How much of a is in the direction of b?”

⸻

📐 Types of Projection

1️⃣ Scalar Projection (length only)

2️⃣ Vector Projection (actual projected vector)


image.png

image.png

```python
# Vector Projection
# Manual + NumPy (Single Block)

import numpy as np

# Define vectors
a = np.array([1, 2])
b = np.array([3, 4])

# -----------------------------
# Step 1: Dot Product
# -----------------------------
dot_product = np.dot(a, b)

# -----------------------------
# Step 2: Magnitude of b
# -----------------------------
magnitude_b = np.linalg.norm(b)

# -----------------------------
# Step 3: Scalar Projection
# -----------------------------
scalar_projection = dot_product / magnitude_b

# -----------------------------
# Step 4: Vector Projection
# -----------------------------
vector_projection = (dot_product / (magnitude_b**2)) * b

# -----------------------------
# Output
# -----------------------------
print("Vector a:", a)
print("Vector b:", b)

print("\nDot Product:", dot_product)
print("Scalar Projection of a onto b:", scalar_projection)
print("Vector Projection of a onto b:", vector_projection)
```

```python

plt.quiver(0, 0, a[0], a[1], angles="xy", scale_units="xy", scale=1, color="blue")
plt.quiver(0, 0, b[0], b[1], angles="xy", scale_units="xy", scale=1, color="orange")
plt.quiver(0, 0, vector_projection[0], vector_projection[1], angles="xy", scale_units="xy", scale=1)
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.title("Vector Projection")
plt.show()
```

# 🔹 Part 3: Line vs Plane vs Hyperplane

Line → 1D object in 2D space

Plane → 2D surface in 3D space

Hyperplane → n-1 dimensional separator in n-dimensional space

Used in Machine Learning as decision boundary.


```python

x = np.linspace(-5, 5, 100)
y = 2*x + 1

plt.plot(x, y)
plt.title("Line in 2D (y = 2x + 1)")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
