# 📐 Linear Algebra – Important Concepts

## 1️⃣ How is Angle Between Vectors Calculated?

The angle (θ) between two vectors **A** and **B** is calculated using the dot product formula:

cosθ = (A · B) / (||A|| ||B||)

Where:
- A · B = Dot product of vectors
- ||A|| = Magnitude of vector A
- ||B|| = Magnitude of vector B

Then,

θ = cos⁻¹ [(A · B) / (||A|| ||B||)]

👉 This formula tells how similar the directions of two vectors are.

---

## 2️⃣ What is Vector Projection?

Vector projection is the process of projecting one vector onto another vector.

It gives the component of one vector in the direction of another vector.

Projection of A onto B:

proj₍B₎ A = [(A · B) / ||B||²] B

👉 It tells how much of vector A lies in the direction of vector B.

---

## 3️⃣ Difference Between Line, Plane, and Hyperplane

| Concept | Description | Dimension |
|----------|------------|------------|
| Line | Straight path extending in two directions | 1D |
| Plane | Flat surface extending infinitely | 2D |
| Hyperplane | Generalization of plane in higher dimensions | n−1 dimension |

Examples:
- In 2D → Line divides plane
- In 3D → Plane divides space
- In n-dimension → Hyperplane divides space

👉 Hyperplane is an extension of plane to higher dimensions.

---

## 4️⃣ Define Matrix

A matrix is a rectangular arrangement of numbers arranged in rows and columns.

Example:

A =  
[ 1  2  3  
  4  5  6 ]

👉 It is used to represent data, systems of equations, and transformations.

### 6️⃣ Difference Between Matrix Multiplication and Element-Wise Multiplication

| Basis | Matrix Multiplication | Element-wise Multiplication |
|--------|----------------------|-----------------------------|
| Rule | Row × Column method | Multiply corresponding elements |
| Size Condition | Columns of A = Rows of B | Matrices must be same size |
| Result | New matrix based on dot products | Each element multiplied directly |
| Used In | Linear algebra transformations | Machine learning / programming |

Example:

Matrix multiplication:
A(2×3) × B(3×2)

Element-wise:
A(2×2) ⊙ B(2×2)

👉 Matrix multiplication follows algebra rules  
👉 Element-wise multiplication is direct position-wise multiplication

```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
# E1=[4,6] 
# E2=[5,8]
E1 = np.array([4, 6])
E2 = np.array([5, 8])
```

```python
dot_product = np.dot(E1, E2)
mag_E1 = np.linalg.norm(E1)
mag_E2 = np.linalg.norm(E2)

cos_theta = dot_product / (mag_E1 * mag_E2)

angle = math.degrees(math.acos(cos_theta))
print("Angle between E1 and E2 (degrees) : ",angle)
```

### Insight

The angle between **E1** and **E2** is approximately **1.68°**, which is very close to 0°.  

This indicates that the two vectors are almost perfectly aligned and point in nearly the same direction.  


```python
projection = (np.dot(E1, E2) / np.dot(E2, E2)) * E2 
print("Projection of E1 and E2 : ",projection)
```

```python
plt.figure(figsize=(6,6))

plt.gca().set_facecolor("#f5f7fa")

# Plot E1
plt.quiver(0, 0, E1[0], E1[1], 
           angles='xy', scale_units='xy', scale=1, color="#1f77b4", width=0.008, label="E1 (Your Skills)")

# Plot E2
plt.quiver(0, 0, E2[0], E2[1], 
           angles='xy', scale_units='xy', scale=1, color="#2ca02c", width=0.008, label="E2 (Industry Skills)")

# Plot Projection
plt.quiver(0, 0, projection[0], projection[1], 
           angles='xy', scale_units='xy', scale=1, color="#d62728", width=0.010, alpha=0.9,
           label="Projection of E1 onto E2")

plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.title("Skill Vector Visualization")
plt.show()
```

### Insight :
- This indicates a strong similarity and high directional match between them.

```python
# Create matrix
Skill_Matrix = np.array([
    [4, 6],
    [5, 8]
])

print("Skill Matrix:\n", Skill_Matrix)
```

```python
determinant = np.linalg.det(Skill_Matrix)
print("Determinant:", determinant) 
```

### Meaning:
- If determinant = 0 → Skills are linearly dependent
- If determinant ≠ 0 → Skills are independent

```python
if determinant != 0:
    inverse = np.linalg.inv(Skill_Matrix)
    print("Inverse Matrix:\n", inverse)
else: 
    print("Matrix is singular (no inverse).")
```

### Hyperplane in ML Context

In Machine Learning (especially Support Vector Machine - SVM):

A hyperplane is defined as: \
w1x1 + w2x2 + w3x3 + ... + wnxn + b = 0
### In Different Dimensions:

- 2D → Line  
- 3D → Plane  
- Higher dimensions → Hyperplane  

### Purpose:

A hyperplane separates two classes with the **maximum margin**.

### Example:

If:
- \( x_1 \) = Coding Skill  
- \( x_2 \) = Communication Skill  

Then the hyperplane separates:

- Hire  
- Not Hire
