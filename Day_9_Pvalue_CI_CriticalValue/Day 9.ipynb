# Question  Q.1
- Generate a sample of 100 heights.
- Calculate the sample mean and standard deviation.
- Compute the 95% confidence interval range.
- Plot the sample with its confidence interval.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
np.random.seed(42)
sample_size = 100
heights = np.random.normal(loc=170, scale=10, size=sample_size)

sample_mean = np.mean(heights)
sample_std = np.std(heights, ddof=1)

confidence_level = 0.95
alpha = 1 - confidence_level
t_critical = stats.t.ppf(1 - alpha/2, df=sample_size - 1)
margin_of_error = t_critical * (sample_std / np.sqrt(sample_size))

ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

plt.figure()
plt.hist(heights, bins=15, alpha=0.7, label="Sample Heights")
plt.axvline(sample_mean, linestyle='--', label=f"Mean = {sample_mean:.2f}")
plt.axvline(ci_lower, linestyle=':', label=f"95% CI Lower = {ci_lower:.2f}")
plt.axvline(ci_upper, linestyle=':', label=f"95% CI Upper = {ci_upper:.2f}")

plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.title("Sample Heights with 95% Confidence Interval")
plt.legend()
plt.show()
```

# Q : 2
- Create a normal distribution curve.
- Shade the area in both tails. (use plt.fill_between() method of matplotlib)
- Explain that this shaded area represents the p-value in a two-tailed test.

```python
from scipy.stats import norm

x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)

z_critical = 1.96

plt.figure()
plt.plot(x, y, label="Normal Distribution")

plt.fill_between(x, y, where=(x <= -z_critical))
plt.fill_between(x, y, where=(x >= z_critical))

plt.axvline(-z_critical, linestyle='--', label="-1.96")
plt.axvline(z_critical, linestyle='--', label="+1.96")

plt.xlabel("Z-value")
plt.ylabel("Probability Density")
plt.title("Two-Tailed Test: Shaded Area Represents p-value")
plt.legend()

plt.show()
```

### Explanation

- In a two-tailed hypothesis test, extreme values can occur on both sides of the distribution.
- The shaded areas in both tails represent the p-value, which is the probability of observing results as extreme as (or more extreme than) the test statistic when the null hypothesis is true.

# Q : 3

- Plot a standard normal distribution.
- Mark critical values for 90%, 95%, and 99% confidence levels.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)

z_90 = 1.645   # 90% confidence level
z_95 = 1.96    # 95% confidence level
z_99 = 2.576   # 99% confidence level

plt.figure()
plt.plot(x, y, label="Standard Normal Distribution")

plt.axvline(-z_90, linestyle='--', label="90% Confidence Level")
plt.axvline(z_90, linestyle='--')

plt.axvline(-z_95, linestyle=':', label="95% Confidence Level")
plt.axvline(z_95, linestyle=':')

plt.axvline(-z_99, linestyle='-.', label="99% Confidence Level")
plt.axvline(z_99, linestyle='-.')

plt.xlabel("Z-value")
plt.ylabel("Probability Density")
plt.title("Standard Normal Distribution with Critical Values")
plt.legend()

plt.show()
```
