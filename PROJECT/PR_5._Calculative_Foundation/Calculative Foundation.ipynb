# Part A: Vector & Matrix Fundamentals

### 1. Represent each student's subject scores as a vector.

```python
import numpy as np

S1 = np.array([85, 78, 92, 88])
S2 = np.array([80, 82, 89, 90])

# Norm-1
norm1_S1 = np.linalg.norm(S1, 1)
print("Norm-1 of S1:", norm1_S1)

# Norm-2
norm2_S1 = np.linalg.norm(S1)
print("Norm-2 of S1:", norm2_S1)
```

### 2. Compute:
- Norm-1 and Norm-2 of vectors.
- Dot product and angle between two students' score vectors.
- Cross product (for 3D selected subjects).

```python
dot_product = np.dot(S1, S2)
print("Dot product of S1 and S2 : ",dot_product)
```

```python
angle = np.arccos(dot_product / (np.linalg.norm(S1) * np.linalg.norm(S2)))
angle_deg = np.degrees(angle)
print("Angle bwtween S1 & S2 (degrees) : ",angle_deg)
```

```python
S1_3D = S1[:3]
S2_3D = S2[:3]
cross_product = np.cross(S1_3D, S2_3D)
print("Cross product of S1 & S2 (3D):", cross_product)
```

### 3. Find the projection of one vector onto another.

```python
projection = (np.dot(S1, S2) / np.linalg.norm(S2)**2) * S2
print("Projection of S1 onto S2:", projection)
```

# Part B: Matrix Operations

### 4. Form a matrix of students subjects. 
**Perform:**
- Matrix addition and multiplication.
- Transpose and Inverse (if possible).
- Determinant.


```python
A = np.array([
    [85, 90, 78],
    [80, 88, 92],
    [75, 85, 89]
])

B = np.array([
    [70, 75, 80],
    [68, 72, 78],
    [74, 76, 82]
])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)
```

```python
# 1. Matrix Addition
addition = A + B
print("\nMatrix Addition (A + B):\n", addition)
```

```python
# 2. Matrix Multiplication
multiplication = np.dot(A, B)
print("\nMatrix Multiplication (A × B):\n", multiplication)
```

```python
# 3. Transpose of Matrix
transpose = A.T
print("\nTranspose of Matrix A:\n", transpose)
```

```python
# 4. Determinant
determinant = np.linalg.det(A)
print("\nDeterminant of Matrix A:", determinant)
```

```python
# 5. Inverse of Matrix

if determinant != 0:
    inverse = np.linalg.inv(A)
    print("\nInverse of Matrix A:\n", inverse)
else:
    print("\nMatrix A is singular. Inverse not possible.")
```

# Part C: Linear Transformations & Geometry

### 5. Explain line, plane, and hyperplane with respect to your dataset dimensions.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

math = np.array([85, 80, 75])
science = np.array([90, 88, 85])
english = np.array([78, 92, 89])

fig = plt.figure(figsize=(15,5))

ax1 = fig.add_subplot(1,3,1)

ax1.scatter(math, [0,0,0])

for i in range(len(math)):
    ax1.text(math[i], 0, f"S{i+1}")

ax1.set_title("1D Line (Math Scores)")
ax1.set_xlabel("Math Score")
ax1.set_yticks([])

ax2 = fig.add_subplot(1,3,2)

ax2.scatter(math, science)

for i in range(len(math)):
    ax2.text(math[i], science[i], f"S{i+1}")

ax2.set_title("2D Plane (Math vs Science)")
ax2.set_xlabel("Math")
ax2.set_ylabel("Science")

ax3 = fig.add_subplot(1,3,3, projection='3d')

ax3.scatter(math, science, english)

for i in range(len(math)):
    ax3.text(math[i], science[i], english[i], f"S{i+1}")

ax3.set_title("3D Feature Space")
ax3.set_xlabel("Math")
ax3.set_ylabel("Science")
ax3.set_zlabel("English")

plt.tight_layout()
plt.show()
```

### Insight :
> The subplot visualization demonstrates how dataset dimensionality increases with the number of features. A single feature forms a line (1D), two features form a plane (2D), and three features create a 3D feature space where hyperplanes can separate data points.

### 6. Show how dimensionality increases from 2D 3D higher dimensions with hyperplanes.

```python
math = np.array([85, 80, 75, 90])
science = np.array([90, 88, 85, 92])
english = np.array([78, 92, 89, 85])
history = np.array([88, 84, 82, 91]) 

fig = plt.figure(figsize=(15,5))

ax1 = fig.add_subplot(1,3,1)

ax1.scatter(math, science)

for i in range(len(math)):
    ax1.text(math[i], science[i], f"S{i+1}")

ax1.set_title("2D Plane (Math vs Science)")
ax1.set_xlabel("Math")
ax1.set_ylabel("Science")

ax2 = fig.add_subplot(1,3,2, projection='3d')

ax2.scatter(math, science, english)

for i in range(len(math)):
    ax2.text(math[i], science[i], english[i], f"S{i+1}")

ax2.set_title("3D Space (Math, Science, English)")
ax2.set_xlabel("Math")
ax2.set_ylabel("Science")
ax2.set_zlabel("English")

ax3 = fig.add_subplot(1,3,3)

ax3.scatter(math, science)

for i in range(len(math)):
    ax3.text(math[i], science[i], f"S{i+1}\nH:{history[i]}")

ax3.set_title("Higher Dimension Concept (4 Features)")
ax3.set_xlabel("Math")
ax3.set_ylabel("Science")

ax3.text(76, 91, "English + History\n= Higher Dimensions\n(Hyperplane)", fontsize=9)

plt.tight_layout()
plt.show()
```

# Part D: Eigenvalues & Decomposition

### 7. Compute the eigenvalues and eigenvectors of the covariance matrix.

```python
import numpy as np
from scipy.linalg import lu

X = np.array([
    [85, 90, 78],
    [80, 88, 92],
    [75, 85, 89],
    [90, 92, 85]
])

print("Dataset Matrix:\n", X)

# Step 1: Compute covariance matrix
cov_matrix = np.cov(X.T)

print("\nCovariance Matrix:\n", cov_matrix)

# Step 2: Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("\nEigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
```

### 8. Perform LU Decomposition of the dataset matrix.

```python
P, L, U = lu(X)

print("\nPermutation Matrix (P):\n", P)
print("\nLower Triangular Matrix (L):\n", L)
print("\nUpper Triangular Matrix (U):\n", U)
```

### 9. Perform Singular Value Decomposition (SVD) and explain its role in dimensionality reduction.

```python
U_svd, S, Vt = np.linalg.svd(X)

print("\nU Matrix:\n", U_svd)
print("\nSingular Values:\n", S)
print("\nV Transpose:\n", Vt)
```

# Part E: Dimensionality Reduction

```python
X = np.array([
    [85, 90, 78],
    [80, 88, 92],
    [75, 85, 89],
    [90, 92, 85],
    [70, 75, 80]
])

y = np.array([1, 1, 0, 1, 0])
```

### 10. Apply Principal Component Analysis (PCA) to reduce the dataset from multiple subjects to 2 dimensions.

```python
from sklearn.decomposition import PCA 
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

print("PCA Transformed Data:\n", X_pca)

# Plot PCA
plt.figure()

plt.scatter(X_pca[:,0], X_pca[:,1])

for i in range(len(X_pca)):
    plt.text(X_pca[i,0], X_pca[i,1], f"S{i+1}")

plt.title("PCA: Reduced to 2 Dimensions")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```

### 11. Apply Linear Discriminant Analysis (LDA) to classify students into "Above Average" and "Below Average" categories.

```python
lda = LinearDiscriminantAnalysis(n_components=1)

X_lda = lda.fit_transform(X, y)

print("\nLDA Transformed Data:\n", X_lda)

# Plot LDA
plt.figure()

plt.scatter(X_lda, [0]*len(X_lda), c=y)

for i in range(len(X_lda)):
    plt.text(X_lda[i], 0, f"S{i+1}")

plt.title("LDA Classification Projection")
plt.xlabel("LDA Component")

plt.yticks([])

plt.show()
```
