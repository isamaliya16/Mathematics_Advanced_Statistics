## 1.Define Probability

Probability is the measure of how likely an event is to happen.

It ranges from **0 to 1**:
- 0 → Impossible event  
- 1 → Certain event  

Formula:

P(Event) = Favorable Outcomes / Total Outcomes  

Example:  
If you toss a coin,

Total outcomes = 2 (Head, Tail)  
Probability of getting Head = 1/2 = 0.5  

👉 So, there is a 50% chance of getting Head.

---

## 2.Difference Between Empirical and Theoretical Probability

| Basis | Empirical Probability | Theoretical Probability |
|--------|----------------------|------------------------|
| Meaning | Based on actual experiment | Based on mathematical calculation |
| Data Source | Observed data | Formula & logic |
| Accuracy | May change with more trials | Fixed value (if assumptions correct) |
| Example | Toss coin 100 times and count heads | Probability of head = 1/2 |

👉 Empirical = Practical  
👉 Theoretical = Mathematical

---

## 3.What Does High Probability Indicate in Business?

High probability in business means:

- High chance of success
- Lower risk
- More predictable outcome
- Better decision confidence

Example:
If a product has 90% probability of selling well,  
👉 Business will invest more confidently.

High probability helps companies in:
- Risk management
- Sales forecasting
- Investment decisions
- Market strategy planning

# 🎯 Problem Statement

- Create website visits data for 30 days and perform probability analysis
Generate 30 days of website visit data
- Compute descriptive statistics
- Calculate probability concepts
- Visualize results
- Interpret results (important for exams & projects)

# Step 1: Create Website Visits Data (30 Days)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


np.random.seed(42)
days = np.arange(1,31)
visits = np.random.randint(80,220,size=30)

data = pd.DataFrame({
    "Day" : days,
    "Website_Visits" : visits
})

data
```

# Step 2: Descriptive Statistics

```python
mean_visits = data["Website_Visits"].mean()
median_visits = data["Website_Visits"].median()
std_dev = data["Website_Visits"].std()
min_visits = data["Website_Visits"].min()
max_visits = data["Website_Visits"].max()

print("Website Visits Statistics (30 Days)")
print("----------------------------------")
print(f"Mean Visits           : {mean_visits:.2f}")
print(f"Median Visits         : {median_visits:.2f}")
print(f"Standard Deviation    : {std_dev:.2f}")
print(f"Minimum Visits        : {min_visits}")
print(f"Maximum Visits        : {max_visits}")

```

# Step 3: Visualize Website Traffic

```python
plt.figure(figsize=(10,6))
plt.plot(data["Day"], data["Website_Visits"], marker='o')
plt.axhline(mean_visits, linestyle='--', label="Average Visits")
plt.xlabel("Day")
plt.ylabel("Website Visits")
plt.title("Website Visits Over 30 Days")
plt.legend()
plt.grid()
plt.show()
```

# Step 4: Probability Analysis (Core Statistics)

## 1. Probability of High Traffic Day

```python
# Identify high traffic days (visits greater than 150)
high_traffic_days = data[data["Website_Visits"] > 150]

num_high_traffic_days = len(high_traffic_days)
total_days = len(data)
prob_high_traffic = num_high_traffic_days / total_days
print("High Traffic Day Probability Analysis")
print("------------------------------------")
print("Condition               : Website Visits > 150")
print("Total Days Observed     :", total_days)
print("High Traffic Days       :", num_high_traffic_days)
print(f"Probability of High Traffic Day : {prob_high_traffic:.3f}")
```

## 2.Probability of Low Traffic Day

```python
# Identify high traffic days (visits greater than 150)
low_traffic_days = data[data["Website_Visits"] < 100]

num_low_traffic_days = len(low_traffic_days)
total_days = len(data)
prob_low_traffic = num_low_traffic_days / total_days
print("Low Traffic Day Probability Analysis")
print("------------------------------------")
print("Condition               : Website Visits < 100")
print("Total Days Observed     :", total_days)
print("Low Traffic Days       :", num_low_traffic_days)
print(f"Probability of Low Traffic Day : {prob_low_traffic:.3f}")
```

## 3.Probability Distribution (Histogram)

```python
plt.figure(figsize=(8,5))
plt.hist(data["Website_Visits"], bins=8, density=True)
plt.xlabel("Website Visits")
plt.ylabel("Probability")
plt.title("Probability Distribution of Website Visits")
plt.show()
```

# Step 5: Normal Probability 

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Mean and standard deviation
mu = mean_visits
sigma = std_dev

# Create x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Normal distribution curve
y = norm.pdf(x, mu, sigma)

# Plot the normal curve
plt.figure(figsize=(10,6))
plt.plot(x, y)
plt.xlabel("Website Visits")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Website Visits")

# Shade area between 120 and 180
x_fill = np.linspace(120, 180, 300)
y_fill = norm.pdf(x_fill, mu, sigma)
plt.fill_between(x_fill, y_fill, alpha=0.3)

# Mark mean line
plt.axvline(mu, linestyle='--', label="Mean")

plt.legend()
plt.grid()
plt.show()

prob_range = norm.cdf(180, mu, sigma) - norm.cdf(120, mu, sigma)
print(f"Probability of visits between 120 and 180: {prob_range:.3f}")
```
