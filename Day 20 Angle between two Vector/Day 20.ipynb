```python
import numpy as np 
import math 
```

```python
a = np.array([1, 2])
b = np.array([3, 4])
dot_product = np.dot(a, b)

mag_a = np.linalg.norm(a)
mag_b = np.linalg.norm(b)

cos_theta = dot_product / (mag_a * mag_b)
angle = math.degrees(math.acos(cos_theta))

print("Angle between a and b (in degrees) : ",angle)
```

```python
v1 = np.random.randint(1, 10, 3)
v2 = np.random.randint(1, 10, 3)

dot_product = np.dot(v1, v2)
mag_v1 = np.linalg.norm(v1)
mag_v2 = np.linalg.norm(v2)

cos_theta = dot_product / (mag_v1 * mag_v2)
angle = math.degrees(math.acos(cos_theta))

print("Vector 1:", v1)
print("Vector 2:", v2)
print("Angle (degrees):", angle)
```

```python
a = np.array([3, 4])
b = np.array([1, 2])
projection = (np.dot(a,b) / np.dot(b,b)) * b 

print("Projection of a on b : ",projection)
```

```python
import matplotlib.pyplot as plt 
plt.figure() 
plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, projection[0], projection[1], angles='xy', scale_units='xy', scale=1)
plt.xlim(0,5)
plt.ylim(0,5)
plt.grid()
plt.title("Vector Projection (2D)")
plt.show()
```

## Q.4
#### Represent a plane equation in 3D: x + y + z = 5 and visualize with matplotlib.

```python
from mpl_toolkits.mplot3d import Axes3D

# Create grid
x = np.linspace(-5, 5, 10)
y = np.linspace(-5, 5, 10)
x, y = np.meshgrid(x, y)

# Compute z
z = 5 - x - y

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z)

ax.set_title("Plane: x + y + z = 5")
plt.show()
```

## Q.5
### Explain how a hyperplane in 4D is used in SVM classification (conceptual task, no code).


## Hyperplane in 4D for SVM Classification (Conceptual Explanation)

- In Support Vector Machine (SVM) classification, a **hyperplane** is a decision boundary used to separate data points belonging to different classes.
---

## 🔹 What is a 4D Hyperplane?

If we have four input features:

- x1  
- x2  
- x3  
- x4  

Then the hyperplane equation is:

w1x1 + w2x2 + w3x3 + w4x4 + b = 0

This equation represents a hyperplane in **4-dimensional feature space**.

---

## 🔹 How It Works in SVM

- Each data point is represented as a point in 4D space:

  (x1, x2, x3, x4)

- The SVM algorithm finds the **optimal hyperplane** that:

  - Separates two classes  
  - Maximizes the margin (distance between the boundary and the nearest points)

- The closest data points to the hyperplane are called:

  **Support Vectors**

---

## 🔹 Classification Rule

- If w · x + b > 0 → Class 1  
- If w · x + b < 0 → Class 2  

---

## 🔹 Why 4D?

In real-world machine learning problems, we often use multiple features.

Example:

- x1 = Technical Skill  
- x2 = Communication Skill  
- x3 = Experience  
- x4 = Problem-Solving Ability  

These four features together form a **4D feature space**.

---

## 🔹 Key Idea

Although humans cannot visualize 4D space directly:

- In 2D → Hyperplane is a line  
- In 3D → Hyperplane is a plane  
- In 4D → Hyperplane is a 3D decision boundary inside 4D space

### Q.7
#### Create a random 3x3 matrix and find its determinant and inverse.

```python
A = np.random.randint(1, 10, (3, 3))
print("Matrix A : \n ",A)

det = np.linalg.det(A)
print("Determinant : ",det)

if det != 0:
    inverse = np.linalg.inv(A)
    print("Inverse : \n ",inverse)
else : 
    print("Matrix is singular (no inverse).")
```
