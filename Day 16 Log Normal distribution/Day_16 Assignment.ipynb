
## 1️⃣ What is a Log-Normal Distribution?

A **Log-Normal distribution** is a probability distribution of a variable whose logarithm follows a Normal distribution.

If:

    Y = log(X)

is normally distributed, then:

    X follows a Log-Normal distribution

### Key Properties:
- Values are always positive (X > 0)
- Right-skewed (long tail on the right)
- Mean > Median
- Common in financial and economic data

---

## 2️⃣ Why Does Income Often Follow Log-Normal Distribution?

Income tends to follow a Log-Normal distribution because:

- Income grows multiplicatively (raises, investments, business profits).
- Many small proportional growth factors combine over time.
- Most people earn moderate income.
- Few individuals earn extremely high income.

Multiplicative growth naturally produces a log-normal shape.

---

## 3️⃣ What is a Power Law Distribution?

A **Power Law distribution** follows the form:

    P(x) ∝ x^(-α)

where:
- α is the power-law exponent
- Large values become less frequent as a power of x

### Key Idea:
Small events are common, but very large events are rare and impactful.

---

## 4️⃣ What Does Heavy-Tailed Mean?

A **heavy-tailed distribution** means:

- Extreme values occur more frequently than in a normal distribution.
- The tail of the distribution decreases slowly.
- Outliers are more common.

Examples:
- Wealth distribution
- Earthquake magnitudes
- Social media followers

Power-law distributions are classic heavy-tailed distributions.

---

## 5️⃣ What is Box-Cox Transformation Used For?

The **Box-Cox transformation** is used to:

- Reduce skewness
- Make data more normally distributed
- Stabilize variance
- Improve statistical modeling

Formula:

For λ ≠ 0:

    Y(λ) = (X^λ − 1) / λ

For λ = 0:

    Y = log(X)

---

## 6️⃣ Why Must Data Be Positive for Box-Cox?

Box-Cox requires strictly positive data because:

- It involves logarithms and power transformations.
- Log(0) and log(negative numbers) are undefined.
- Raising negative values to fractional powers may produce complex numbers.

Therefore:

    X > 0  is required.

---

## 7️⃣ How Does Box-Cox Help in Regression?

Box-Cox improves regression by:

- Making residuals more normally distributed
- Reducing heteroscedasticity (non-constant variance)
- Improving linear relationship
- Increasing model accuracy
- Helping satisfy regression assumptions

Result:
Better model performance and more reliable predictions.

# 1. Import Required Libraries

```python
import numpy as np 
import matplotlib.pyplot as plt 
import scipy.stats as stats 
from scipy.stats import boxcox 
```

# 2. Simulate Log-Normal Revenue Data

```python
np.random.seed(42)
mu = 3.5 
sigma = 1.2
revenue = np.random.lognormal(mu,sigma,5000)
```

# 3. Plot Histogram of Revenue

```python
plt.figure()
plt.hist(revenue, bins = 50)
plt.title("Histogram of Log-Normal Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

#### I Observe :
- Heavy right skew
- Few sellers earning extremely high revenue

# 4. Apply Box-Cox Transformation
- Box-Cox helps reduce skewness and normalize data.

```python
revenue_boxcox, lambda_value = boxcox(revenue)
print("Lamda Value : ",lambda_value)
```

# 5. Q-Q Plot Comparison 

```python
plt.figure()
stats.probplot(revenue, dist='norm',plot=plt)
plt.title("Q-Q Plot before Box-Cox")
plt.show()
```

```python
plt.figure()
stats.probplot(revenue_boxcox, dist="norm", plot=plt)
plt.title("Q-Q plot After Box-Cox")
plt.show()
```

### I Observe :
- After Box-Cox:
- Points align closer to straight line
- Data becomes more normal

# 6. Simulate Power-Law Distribution (Follower Count)
- Many platforms follow power-law for followers.

```python
alpha = 2.5 
power_law = (np.random.pareto(alpha,5000) + 1) * 100 
```

# 7. Visual Comparison (Log-Normal vs Power-Law)

```python
plt.figure()
plt.hist(revenue, bins=50, alpha=0.6, density=True)
plt.hist(power_law, bins=50, alpha=0.6, density=True)
plt.title("Log_normal vs Power-Law Distribution")
plt.legend(["Log-Normal Revenue", "Power-Law Followers"])
plt.show()
```

# 📊 Visual Interpretation

| Distribution | Shape | Tail Behavior |
|--------------|--------|---------------|
| Log-Normal   | Right Skewed | Moderately heavy tail |
| Power-Law    | Extremely skewed | Very heavy tail |

Power-law tail decays slower → more extreme values compared to log-normal distribution.
