# Q.2
- Plot them on a 2D plane.
- Show how increasing dimensions represent higher-dimensional points (e.g., 3D point).


```python
import matplotlib.pyplot as plt 

point1 = np.array([2, 3])
point2 = np.array([5, 7])
plt.figure()
plt.scatter(point1[0],point1[1])
plt.scatter(point2[0],point2[1])

plt.text(point1[0], point1[1], "(2,3)")
plt.text(point2[0], point2[1], "(5,7)")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2D Points Representation")
plt.grid(True)
plt.show()
```

```python
point3D = np.array([2, 3, 4])
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

ax.scatter(point3D[0], point3D[1], point3D[2])
ax.text(point3D[0], point3D[1], point3D[2], "(2,3,4)")

ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")
ax.set_title("3D Point Representation")

plt.show()
```

# Q : 5
- Find the Magnitude (Length) of the vector (3,4).

```python
# |v| = √(x² + y²) 
# |v| = √(3² + 4²)
# = √(9 + 16)
# = √25
# = 5

v = np.array([3, 4])
magnitude = np.linalg.norm(v)

print("Vector : ",v)
print("Magnitude (Length) : ", magnitude)
```

# Q.6
- Find the Distance between two points (Euclidean distance).

```python
point1 = np.array([2, 3])
point2 = np.array([5, 7])

# Calculate Euclidean Distance
# d = √[(x₂ − x₁)² + (y₂ − y₁)²]
# d = √[(5−2)² + (7−3)²]
#   = √[(3)² + (4)²]
#   = √(9 + 16)
#   = √25
#   = 5 

distance = np.linalg.norm(point1 - point2)
print("Point 1:", point1)
print("Point 2:", point2)
print("Euclidean Distance:", distance)
```

# Q : 7
- Plot at least 5 random 2D points and calculate distances between them.

```python
np.random.seed(42)
points = np.random.randint(0,10,size=(5,2))
print("Random 2D Points : \n ", points) 
```

```python
plt.figure()

for i in range(5):
    x = points[i][0]
    y = points[i][1]
    plt.scatter(x, y)
    plt.text(x, y, f"P{i}")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("5 Random 2D Points")
plt.grid(True)
plt.show() 
```

```python
print("Euclidean Distances Between Consecutive Points:\n")

for i in range(len(points) - 1):
    distance = np.linalg.norm(points[i+1] - points[i])
    print(f"Distance between P{i} and P{i+1}: {round(distance, 2)}")
```
