# 📊 ANOVA (Analysis of Variance)

## ✅ When to Use an ANOVA Test?

ANOVA is used when you want to determine whether **three or more group means** differ significantly.

Use ANOVA when:

- You are comparing **multiple groups**
- The dependent variable is **numerical**
- Samples are **independent**
- Data is approximately **normally distributed**
- Group variances are roughly **equal**

👉 Instead of running many t-tests (which increases error), ANOVA performs a **single overall comparison**.

---

## ✅ Types of ANOVA Tests

### 🔹 1. One-Way ANOVA

**Purpose:**  
Compares means of groups based on **one independent factor**.

**Example:** Comparing exam scores across different classes.

**Formula:**

### F = MS_between / MS_within  
Where: 
- MS_between = SS_between / (k − 1)  
- MS_within = SS_within / (N − k)
---

### 🔹 2. Two-Way ANOVA

**Purpose:**  
Evaluates the effect of **two independent factors** and their interaction.

**Example:** Teaching method × study time on performance.

**Model Formula:**
X = μ + A + B + AB + error

Where:
- A → Factor A effect  
- B → Factor B effect  
- AB → Interaction effect  
---

### 🔹 3. Repeated Measures ANOVA

**Purpose:**  
Used when the **same subjects** are measured multiple times.

**Example:** Measuring weight loss over several months.

**Formula Structure:**

F = Variance_between_treatments / Variance_within_subjects

This controls individual variability.

---

### ✅ Difference Between One-Way and Two-Way ANOVA

| Feature | One-Way ANOVA | Two-Way ANOVA |
|--------|---------------|---------------|
| Factors analyzed | One | Two |
| Purpose | Compare group means | Compare means + interaction |
| Complexity | Simple | More advanced |
| Interaction study | Not possible | Possible |
| Example | Scores by class | Scores by class & method |

---


### A company tests 3 marketing campaigns.

- Campaign A: [45, 48, 50, 47]
- Campaign B: [52, 55, 53, 54]
- Campaign C: [60, 58, 62, 59]

#### Tasks:

1. State H₀ and H₁
2. Perform manual ANOVA
3. Use f_oneway()
4. Compute critical F-value
5. Apply both decision methods
Interpret business meaning

```python
from scipy.stats import f  
 
A = [45, 48, 50, 47]
B = [52, 55, 53, 54]
C = [60, 58, 62, 59]

groups = [A, B, C]
alpha = 0.05

print("=== Hypotheses ===")
print("H0: All campaign means are equal")
print("H1: At least one campaign mean differs\n")
 
# STEP 1 — Sums & Means 
print("STEP 1 — Group Means")

means = []
sums = []

for i, g in enumerate(groups, start=1):
    s = sum(g)
    m = s / len(g)
    sums.append(s)
    means.append(m)

    print(f"Campaign {i} sum =", s)
    print(f"Campaign {i} mean =", m, "\n")
 
# STEP 2 — Grand Mean 
total_sum = sum(sums)
N = sum(len(g) for g in groups)

grand_mean = total_sum / N

print("STEP 2 — Grand Mean =", round(grand_mean, 3), "\n")
 
# STEP 3 — Between-group SS 
SSB = 0

for g, m in zip(groups, means):
    SSB += len(g) * (m - grand_mean) ** 2

print("STEP 3 — SSB =", round(SSB, 3))
 
# STEP 4 — Within-group SS 
SSW = 0

for g, m in zip(groups, means):
    for x in g:
        SSW += (x - m) ** 2

print("STEP 4 — SSW =", round(SSW, 3), "\n")
 
# STEP 5 — Degrees of Freedom 
k = len(groups)
df_between = k - 1
df_within = N - k
print("STEP 5 — Degrees of Freedom")
print("df_between =", df_between)
print("df_within =", df_within, "\n")
 
# STEP 6 — Mean Squares 
MSB = SSB / df_between
MSW = SSW / df_within

print("STEP 6 — Mean Squares")
print("MSB =", round(MSB, 3))
print("MSW =", round(MSW, 3), "\n")

# STEP 7 — F-statistic
F_stat = MSB / MSW
print("STEP 7 — F-statistic =", round(F_stat, 3))
 
# STEP 8 — Critical Value 
F_critical = f.ppf(1 - alpha, df_between, df_within)
print("Critical F-value =", round(F_critical, 3), "\n")
 
# STEP 9 — Decision 
print("=== Final Decision ===")
if F_stat > F_critical:
    print("Reject H0 → Campaign performances differ significantly.")
else:
    print("Fail to Reject H0 → No significant difference.")
```
