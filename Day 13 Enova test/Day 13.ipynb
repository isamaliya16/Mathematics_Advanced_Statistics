### **Objective:** Check if the average exam scores differ significantly across three different classes.

#### Task Steps:
- Generate three datasets representing scores from Class A, Class B, and Class C.
- Assume each class has 25 students.
- Perform a one-way ANOVA test.
- Report the F-statistic and p-value.

```python
classA = [60,65,70,75,80]
classB = [70,75,80,85,90]
classC = [80,85,90,95,100]
alpha = 0.05

print("=== Hypotheses for One-Way ANOVA ===\n")

print("H0 (Null Hypothesis):")
print("The average exam scores of Class A, Class B, and Class C are equal.")
print("H0: μA = μB = μC\n")

print("H1 (Alternative Hypothesis):")
print("At least one class has a different average exam score.")
print("H1: Not all class means are equal")

# STEP 1 — Calculate sums
print("STEP 1 — Group Sums")
groups = [classA, classB, classC]
sums = []
for i, g in enumerate(groups, start=1):
    s = sum(g)
    sums.append(s)
    print(f"Class {i} sum : ",s)

# STEP 2 — Calculate means
print("\nSTEP 2 — Group Means")
means = []
for i, g in enumerate(groups,start=1):
    m = sum(g) / len(g)
    means.append(m)
    print(f"class {i} mean : ",m)
    
# STEP 3 — Grand mean
print("\nSTEP 3 — Grand Mean")
total_sum = sum(sums)
total_n = sum(len(g) for g in groups)
grand_mean = total_sum / total_n 
print("Grand mean : ",grand_mean)

# STEP 4 — Between-group variation (SSB)
print("\nSTEP 4 — Between Group Sum of Squares (SSB)")
SSB = 0 
for g, mean in zip(groups, means) : 
    SSB += len(g) * (mean - grand_mean)**2
print("SSB : ",SSB)

# STEP 5 — Within-group variation (SSW)
print("\nSTEP 5 — Within Group Sum of Squares (SSW)")
SSW = 0 
for g, mean in zip(groups, means):
    for x in g: 
        SSW += (x - mean)**2 
print("SSW : ",SSW)

# STEP 6 — Degrees of freedom
print("\nSTEP 6 — Degrees of Freedom")
k = len(groups)
N = total_n 
df_between = k - 1
df_within = N - k 
print("df_between =", df_between)
print("df_within =", df_within)

# STEP 7 — Mean squares
print("\nSTEP 7 — Mean Squares")
MSB = SSB / df_between 
MSW = SSW / df_within  
print("MSB =", MSB)
print("MSW =", MSW)

# STEP 8 — F-statistic
print("\nSTEP 8 — F-statistic")
F = MSB / MSW
print("F-statistic =", F )

# STEP 9 — Critical value
print("\nSTEP 9 — Critical Value")
F_critical = f.ppf(1 - alpha, df_between, df_within)
print("Critical F-value =", round(F_critical, 4))

print("\n=== Final Decision ===")
if F_stat > F_critical:
    print("Reject H0 → Significant difference exists.")
else:
    print("Fail to Reject H0 → No significant difference.")
```

# Q : 2
## Plant Growth Under Different Fertilizers
**Objective:** Test if different types of fertilizers have different effects on plant growth.
#### **Task Steps :**
1. Create growth measurements for plants under Fertilizer A, . Fertilizer B, and Fertilizer C.
3. Each group should have 20 plant heights.
4. Perform a one-way ANOVA test.
5. Report the F-statistic and p-value.


```python
import numpy as np
from scipy.stats import f_oneway

# For reproducibility
np.random.seed(10)

# Step 1 — Generate plant growth data (cm)
fert_A = np.random.normal(20, 2, 20)
fert_B = np.random.normal(23, 2, 20)
fert_C = np.random.normal(26, 2, 20)

print("Mean growth under Fertilizers:")
print("A:", round(np.mean(fert_A), 2))
print("B:", round(np.mean(fert_B), 2))
print("C:", round(np.mean(fert_C), 2))

# Step 2 — Perform one-way ANOVA
F_stat, p_value = f_oneway(fert_A, fert_B, fert_C)

# Step 3 — Report results
print("\n=== ANOVA Result ===")
print("F-statistic =", round(F_stat, 4))
print("p-value =", round(p_value, 6))

# Step 4 — Decision
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject H0")
    print("Different fertilizers significantly affect plant growth.")
else:
    print("\nConclusion: Fail to Reject H0")
    print("No significant difference in fertilizer effects.")
```

# Q : 3
Sales Performance Across Regions
**Objective:** Determine if there is a significant difference in average monthly sales across three regions.
- **Task Steps:** 
1. Generate monthly sales data for Region North, South, and West (30 entries each).
2. Perform a one-way ANOVA.
3. Print the F-statistic and p-value.
4. Comment if regional differences are statistically notable.



```python
import numpy as np
from scipy.stats import f_oneway, f

# Step 1 — Generate sales data
np.random.seed(5)

north = np.random.normal(500, 40, 30)
south = np.random.normal(540, 40, 30)
west  = np.random.normal(580, 40, 30)

# Step 2 — Perform one-way ANOVA
F_stat, p_value = f_oneway(north, south, west)

# Step 3 — Degrees of freedom
k = 3
N = 90

df_between = k - 1
df_within = N - k

# Step 4 — Critical value
alpha = 0.05
F_critical = f.ppf(1 - alpha, df_between, df_within)

# Step 5 — Display results
print("=== Sales Performance ANOVA ===\n")

print("Mean Sales:")
print("North:", round(np.mean(north), 2))
print("South:", round(np.mean(south), 2))
print("West :", round(np.mean(west), 2))

print("\nF-statistic =", round(F_stat, 4))
print("p-value =", round(p_value, 6))
print("Critical F-value =", round(F_critical, 4))

# Step 6 — Decision (p-value method)
print("\nDecision using p-value:")
if p_value < alpha:
    print("Reject H0 → Regional sales differ significantly.")
else:
    print("Fail to Reject H0 → No significant difference.")

# Step 7 — Decision (critical value method)
print("\nDecision using critical value:")
if F_stat > F_critical:
    print("Reject H0 → Regional differences are statistically notable.")
else:
    print("Fail to Reject H0 → Differences are not significant.")
```

# Q.4
### Comparing Average App Usage Time Across Age Groups
**Objective:** Evaluate whether app usage time differs among three age groups.
- **Task Steps:**
1. Create datasets for usage time (in minutes) for Teens, Adults, and Seniors.
2. Each group should have 30 samples.
3. Conduct a one-way ANOVA.
4. Report the F-statistic and p-value.

```python
import numpy as np
from scipy.stats import f_oneway

# Step 1 — Generate datasets (minutes)
np.random.seed(7)

teens   = np.random.normal(180, 25, 30)
adults  = np.random.normal(140, 20, 30)
seniors = np.random.normal(90, 15, 30)

# Display group means
print("=== Average Usage Time (minutes) ===")
print("Teens  :", round(np.mean(teens), 2))
print("Adults :", round(np.mean(adults), 2))
print("Seniors:", round(np.mean(seniors), 2))

# Step 2 — Perform one-way ANOVA
F_stat, p_value = f_oneway(teens, adults, seniors)

# Step 3 — Report results
print("\n=== ANOVA Result ===")
print("F-statistic =", round(F_stat, 4))
print("p-value =", round(p_value, 6))

# Step 4 — Interpretation
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject H0")
    print("App usage time differs significantly among age groups.")
else:
    print("\nConclusion: Fail to Reject H0")
    print("No significant difference in usage time.")
```
