## 📊 Statistics: Descriptive vs Inferential

Statistics is broadly divided into **two main types** based on how data is used and interpreted.

---

## 🔹 Descriptive Statistics

**Descriptive Statistics** focuses on **summarizing and organizing data** to understand its main characteristics.  
It does **not** make predictions or conclusions beyond the given data.

### ✅ Purpose
- To **describe** what the data shows  
- To simplify large datasets into meaningful insights  

### 📌 Common Techniques
- **Measures of Central Tendency**
  - Mean  
  - Median  
  - Mode  
- **Measures of Dispersion**
  - Range  
  - Variance  
  - Standard Deviation  
- **Data Visualization**
  - Bar charts  
  - Histograms  
  - Pie charts  
  - Line graphs  

### 📘 Example
> Calculating the **average monthly sales** of a company using past sales data.

---

## 🔹 Inferential Statistics

**Inferential Statistics** uses **sample data** to make **predictions, estimates, or decisions** about a larger population.

### ✅ Purpose
- To **draw conclusions** beyond the observed data  
- To support **decision-making and forecasting**  

### 📌 Common Techniques
- **Probability Distributions**  
- **Hypothesis Testing**
  - t-test  
  - z-test  
  - chi-square test  
- **Estimation**
  - Confidence intervals  
- **Regression & Correlation Analysis**  

### 📘 Example
> Predicting **future sales trends** based on a sample of past sales data.

---


```python
import pandas as pd

# Create a sample dataset
data = {
    'Sales': [120, 135, 150, 145, 160, 155, 170, 165, 180, 175],
    'Profit': [20, 25, 30, 28, 35, 33, 40, 38, 45, 42]
}

# Load data into DataFrame
df = pd.DataFrame(data)

# Display dataset
print(df)
```

```python
summary = df.describe()
print(summary)
```

## 📊 Values Returned by the `.describe()` Method and Their Meaning

The Pandas `.describe()` method returns a set of **summary statistics** for numerical columns.  
Each value helps in understanding the **distribution, spread, and range** of the data.

---

### 🔹 1. `count`
- Represents the **number of non-null (valid) values**
- Useful for identifying **missing data**

**Example:**  
If `count = 100`, there are 100 valid observations.

---

### 🔹 2. `mean`
- The **average value** of the dataset
- Calculated as:
  
**Example:**  
Average sales across all months.

---

### 🔹 3. `std` (Standard Deviation)
- Measures the **spread or variability** of the data
- Higher value → more variation  
- Lower value → values are closer to the mean

**Example:**  
How much sales values differ from the average sales.

---

### 🔹 4. `min`
- The **smallest value** in the dataset

**Example:**  
Lowest sales or profit recorded.

---

### 🔹 5. `25%` (First Quartile / Q1)
- 25% of the values are **below this number**
- Represents the **lower quarter** of the data

**Example:**  
25% of sales values fall below this point.

---

### 🔹 6. `50%` (Median / Q2)
- The **middle value** of the dataset
- Half of the values are below and half are above

**Example:**  
Typical or central sales value.

---

### 🔹 7. `75%` (Third Quartile / Q3)
- 75% of the values are **below this number**
- Shows the **upper spread** of the data

**Example:**  
Most sales values are less than or equal to this value.

---

### 🔹 8. `max`
- The **largest value** in the dataset

**Example:**  
Highest sales or profit recorded.

---

## 📌 Summary Table

| Statistic | What It Tells |
|--------|---------------|
| count | Number of valid values |
| mean | Average value |
| std | Data variability |
| min | Lowest value |
| 25% | Lower quartile |
| 50% | Median |
| 75% | Upper quartile |
| max | Highest value |

---

## ✨ Simple Explanation

> The `.describe()` method tells **how many values exist**, **how large or small they are**, and **how they are spread across the dataset**.


## Mean

```python
# mean using pandas
mean_sales_pandas = df['Sales'].mean()
print("Mean : ",mean_sales_pandas)
```

### 🔍 What happens?
- Pandas automatically ignores missing values (NaN)
- Simple and most commonly used in data analysis

```python
# mean using numpy 
import numpy as np 
mean_sales_numpy = np.mean(df['Sales'])
print("Mean using Numpy : ",mean_sales_numpy)
```

### 🔍 What happens?

- NumPy calculates the arithmetic average
- If NaN values exist, it will return NaN
    (use np.nanmean() to ignore NaNs)

## Medain

```python
# medain using pandas
median_sales_pandas = df['Sales'].median()
print("Median using pandas : ",median_sales_pandas)
```

### 🔍 Explanation

- Pandas automatically ignores NaN values
- Returns the middle value of the sorted data

```python
# medain using numpy
median_sales_numpy = np.median(df['Sales'])
print("Median using Numpy : ",median_sales_numpy)
```

### 🔍 Explanation
- NumPy calculates the median mathematically
- If NaN values exist, result will be NaN
- To ignore NaN values, use np.nanmedian()

## Mode 

```python
mode_sales = df['Sales'].mode()
print("Mode Of Sales : ",mode_sales[0])
```

### Important Notes
- .mode() returns a Series, not a single value
- If multiple values have the same highest frequency, all are returned

```python
import matplotlib.pyplot as plt 
mean = df['Sales'].mean()
median = df['Sales'].median()
mode = df['Sales'].mode()[0]  # first mode

measures = ['Mean','Median','Mode']
values = [mean,median,mode]

plt.figure()
plt.bar(measures,values)
plt.title("Bar chart Comparison of mean, median, mode")
plt.xlabel("Statistical Measure")
plt.ylabel("Sales Value")
plt.show()
```

```python
range_sales = df['Sales'].max() - df['Sales'].min()
print("Range of Sales : ",range_sales)
```

### 🔍 Explanation
- Max = 180
- Min = 120
- Range = 180 − 120 = 60
- Shows the spread of the data


## Variance 

```python
# using pandas 
variance_pandas = df['Sales'].var()
print("Variance using Pandas : ",round(variance_pandas,2)) 
```

### 🔍 Important
- Pandas uses sample variance by default
- Internally uses ddof = 1

```python
# using numpy 
variance_numpy = np.var(df['Sales'],ddof=1)
print("Variance using Numpy : ",round(variance_numpy,2))
```

#### ✔️ Both Pandas and NumPy now give the same result.

### 📌 What is `ddof = 1`? 

`ddof = 1` is used to calculate **sample variance or standard deviation** by dividing by **`n − 1`** instead of `n`.  
It corrects bias when estimating population statistics from sample data and is the default in Pandas.


###  Why `ddof = 1` is Important ❗

| ddof value | Meaning |
|----------|---------|
| `ddof = 0` | Population variance |
| `ddof = 1` | Sample variance (used in statistics) |

📌 **Note:**  
Pandas automatically assumes that the data represents a **sample**, not the entire population.  
Therefore, it calculates variance using **`ddof = 1`** by default.

ddof (Delta Degrees of Freedom) is the number subtracted from total observations while calculating variance or standard deviation.
ddof = 1 is used for sample data to reduce estimation bias.

## Standard Deviation 

```python
std_pandas = df['Sales'].std()
print("Standard Deviation using Pandas:", round(std_pandas,2))
```

### 🔍 Explanation

- Pandas calculates sample standard deviation
- Uses ddof = 1 by default
- Measures how far values spread from the mean

```python
std_numpy = np.std(df['Sales'],ddof=1)
print("Standard Deviation using Numpy : ",round(std_numpy,2))
```

### Box plot

```python
plt.figure()
plt.boxplot(df['Sales'])
plt.title("Boxplot of Sales")
plt.ylabel("Sales Value")
plt.show()
```

### 📦 How Dispersion Appears Visually in a Boxplot

#### 🔹 Box (IQR)
- Represents the **Interquartile Range (Q1 to Q3)**
- A **wider box** indicates **higher dispersion**
- A **narrow box** indicates **low dispersion**

#### 🔹 Median Line
- The line inside the box represents the **median**
- If the median is **not centered**, the data is **skewed**

#### 🔹 Whiskers
- Show the spread of data **outside the quartiles**
- **Longer whiskers** indicate **more variability**

#### 🔹 Outliers (if any)
- Points that lie **outside the whiskers**
- Indicate **extreme or unusual values**
