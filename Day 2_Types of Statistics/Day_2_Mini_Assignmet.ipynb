# Monthly Sales Statistics 

## 1. Dataset Creation 

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 

months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales = [120,135,150,145,160,155,170,135,180,175,190,185]

df = pd.DataFrame({
    'Month' : months,
    'Sales' : sales
})
df
```

## 2.  Mean, Median, Mode

```python
import statistics
mean_sales = df['Sales'].mean()
median_sales = df['Sales'].median()
mode_sales = df['Sales'].mode()

print("Mean   : ",round(mean_sales,2))
print("Median : ",round(median_sales,2))
print("Mode   : ",round(mode_sales[0],2))
```

#### 🔍 Insight

- Mean shows average monthly sales.
- Median represents the middle value, less affected by extremes.
- Mode is the most frequent sales value.

## 3. Range,Variance, Standard Deviation 

```python
sales_range = df['Sales'].max() - df['Sales'].min()
variance_sales = df['Sales'].var()
std_sales = df['Sales'].std()

print("Sales Range :",sales_range)
print("Variance  :",round(variance_sales,2))
print("Standard Deviation  :",round(std_sales,2))
```

#### 🔍 Insight

- Range shows spread between highest and lowest sales.
- Variance & SD measure sales consistency.

## 4. Adding an Outlier

```python
# Add an extreme outlier
sales_with_outlier = sales + [4000]

mean_outlier = np.mean(sales_with_outlier)
median_outlier = np.median(sales_with_outlier)

print("Mean   : ",round(mean_outlier,2))
print("Median : ",round(median_outlier,2))
```

#### 🔍 Observation

- Mean and SD increase significantly due to the outlier.
- Median remains relatively stable.

## 5. Histogram Visulization  

```python
plt.figure(figsize=(8,5))
plt.hist(df['Sales'],bins = 6)
plt.title("Monthly Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

#### 📊 Insight

- Histogram shows sales concentration in higher ranges.
- Data is slightly right-skewed.

#### ✅ Final Conclusion

- Monthly sales show a steady increasing trend.
- Outliers strongly affect mean and standard deviation.
- Median is more reliable when extreme values exist.
