# Q : 1
Objective : To check if the sample mean is significantly difference from a known popluation mean.
### **Task Steps:**
- Generate a sample of 50 weights .
- Assume population mean = 70 kg and population standard deviation = 8 kg.
- Perform a one-sample z-test.
- Print the z-value and p-value


![Task 1](images/Q1_1.png)
![Task 1](images/Q1_2.png)
![Task 1](images/Q1_3.png)
![Task 1](images/Q1_4.png)

```python
import numpy as np 
from scipy.stats import norm 

population_mean = 70 
population_std = 8 
n = 50

np.random.seed(42)
sample = np.random.normal(loc=72, scale=8, size=n)

# sample_mean = np.mean(sample)
sample_mean = 72
z_value = (sample_mean - population_mean) / (population_std / np.sqrt(n))

p_value = 2 * (1 - norm.cdf(abs(z_value)))

print("Sample Mean:", round(sample_mean, 3))
print("Z-value:", round(z_value, 3))
print("P-value:", round(p_value, 5))
alpha = 0.05

if p_value < alpha:
    print("Conclusion: Reject Null Hypothesis (Significant Difference)")
else:
    print("Conclusion: Fail to Reject Null Hypothesis (No Significant Difference)")
```

# Q : 2
Objective : To computer means of two independent samples with known population standard deviation. 
### **Task Steps :**
- Generate two samples : Male heights and Female heights.
- Assume both populations have known std
- Perform a two-sample z-test .
- print z-value and p-value .

!["Task : 2"](images/Q2_1.png)


!["Task : 2"](images/Q2_2.png)



!["Task : 2"](images/Q2_3.png)





!["Task : 2"](images/Q2_4.png)



!["Task : 2"](images/Q2_5.png)


```python

n1 = 50   # Sample size (Male)
n2 = 50   # Sample size (Female)

sigma1 = 6  
sigma2 = 5   

np.random.seed(42)  

male_heights = np.random.normal(loc=170, scale=sigma1, size=n1)
female_heights = np.random.normal(loc=162, scale=sigma2, size=n2)

mean1 = np.mean(male_heights)
mean2 = np.mean(female_heights)

# SE = sqrt((σ1² / n1) + (σ2² / n2))
# -----------------------------------
standard_error = np.sqrt((sigma1**2 / n1) + (sigma2**2 / n2))

# Z = (x̄1 - x̄2) / SE
# -----------------------------------
z_value = (mean1 - mean2) / standard_error

p_value = 2 * (1 - norm.cdf(abs(z_value)))

print("Male Mean Height:", round(mean1, 2))
print("Female Mean Height:", round(mean2, 2))
print("Z-value:", round(z_value, 3))
print("P-value:", round(p_value, 6))

alpha = 0.05

if p_value < alpha:
    print("Conclusion: Reject Null Hypothesis (Means are significantly different)")
else:
    print("Conclusion: Fail to Reject Null Hypothesis (No significant difference)")
```

# Q : 3
Objective: Test if the average daily water intake of students differs from the recommended 2 liters/day.

### **Task Steps :**
- Generate a sample dataset of 40 students' daily water intake in liters.
- Assume population mean = 2 liters and population standard deviation = 0.4.
- Perform a one-sample z-test.
- Report the z-score and p-value.

![Task : 3](images/Q3_1.png)



![Task : 3](images/Q3_2.png)




![Task : 3](images/Q3_3.png)


```python
population_mean = 2        # μ (recommended intake in liters)
population_std = 0.4       # σ
n = 40                     # sample size
alpha = 0.05               # significance level

np.random.seed(42)  # for reproducibility
sample = np.random.normal(loc=2.1, scale=0.4, size=n)

sample_mean = np.mean(sample)
# SE = σ / √n
standard_error = population_std / np.sqrt(n)

# Z = (x̄ - μ) / SE
z_score = (sample_mean - population_mean) / standard_error

p_value = 2 * (1 - norm.cdf(abs(z_score)))

print("Sample Mean:", round(sample_mean, 3))
print("Standard Error:", round(standard_error, 4))
print("Z-score:", round(z_score, 3))
print("P-value:", round(p_value, 5))

if p_value < alpha:
    print("Conclusion: Reject Null Hypothesis (Significant difference)")
else:
    print("Conclusion: Fail to Reject Null Hypothesis (No significant difference)")
```

# Q : 4
**Objective:** Compare the average scores of two batches (Batch A and Batch B) of students in the same subject.

### **Task Steps :**
- Generate two datasets of 45 and 50 student scores respectively.
- Assume known population standard deviations for both groups.
- Perform a two-sample z-test.
- Print the test statistic and interpret the result.

![Task : 4](images/Q4_1.png)



![Task : 4](images/Q4_2.png)



![Task : 4](images/Q4_3.png)




![Task : 4](images/Q4_4.png)


```python
n1 = 45
n2 = 50

sigma1 = 10
sigma2 = 12
alpha = 0.05 
np.random.seed(42)

batch_A = np.random.normal(loc=75, scale=sigma1, size=n1)
batch_B = np.random.normal(loc=72, scale=sigma2, size=n2)

mean1 = np.mean(batch_A)
mean2 = np.mean(batch_B)

# SE = sqrt((σ1²/n1) + (σ2²/n2))
standard_error = np.sqrt(sigma1**2 / n1) + (sigma2**2 / n2)

# Z = (x̄1 - x̄2) / SE
z_stat = (mean1 - mean2) / standard_error

p_value = 2 * (1 - norm.cdf(abs(z_stat)))

print("Batch A Mean:", round(mean1, 2))
print("Batch B Mean:", round(mean2, 2))
print("Z-Statistic:", round(z_stat, 3))
print("P-Value:", round(p_value, 5))


if p_value < alpha:
    print("Conclusion: Reject Null Hypothesis")
    print("There is a significant difference between the two batches.")
else:
    print("Conclusion: Fail to Reject Null Hypothesis")
    print("There is no significant difference between the two batches.")
```
