## 1️⃣ When is Poisson Distribution Appropriate?

The **Poisson distribution** is used when:

- We count the number of events in a fixed interval (time, area, distance, etc.)
- Events occur independently
- The average rate (λ) is constant
- Events are rare
- Two events cannot occur at exactly the same time

### Examples:
- Number of customer arrivals per hour
- Number of website failures per day
- Number of phone calls per minute

---

## 2️⃣ What does λ (Lambda) represent?

In a Poisson distribution:

λ = Average number of events per interval

Important properties:

Mean = λ  
Variance = λ  

So, in Poisson distribution:

Mean = Variance = λ

---

## 3️⃣ Define Z-Score

A **Z-score** tells how many standard deviations a value is from the mean.

### Formula:

Z = (X − μ) / σ

Where:
- X = observed value
- μ = mean
- σ = standard deviation

### Interpretation:
- Z = 0 → value equals the mean
- Z > 0 → above the mean
- Z < 0 → below the mean

---

## 4️⃣ Difference Between PDF and CDF

| Feature | PDF (Probability Density Function) | CDF (Cumulative Distribution Function) |
|---------|-------------------------------------|-----------------------------------------|
| Meaning | Density at a point | Probability up to a value |
| Used For | Continuous distributions | Continuous & discrete distributions |
| Range | Area under curve = 1 | Always between 0 and 1 |
| Formula Relation | PDF is derivative of CDF | CDF is integral of PDF |

Mathematical Relation:

CDF(x) = ∫ from -∞ to x of PDF(t) dt

---

## 5️⃣ How to Compute Probability Greater Than a Value Using CDF?

To compute:

P(X > a)

We use:

P(X > a) = 1 − CDF(a)

### Example:

If CDF(10) = 0.75

Then:

P(X > 10) = 1 − 0.75  
P(X > 10) = 0.25

### 1. A hospital receives 8 emergency cases per hour.
- Find probability of exactly 10 cases.
- Plot Poisson distribution.

```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.stats import poisson, norm 

lam = 8
k = 10 

prob_10 = poisson.pmf(k, lam)
print("1. Poisson Distribution")
print("Probability of exactly 10 cases : ",round(prob_10, 6))
x = np.arange(0, 20) 
poisson_probs = poisson.pmf(x, lam)
plt.figure()
plt.stem(x, poisson_probs)
plt.title("Poisson Distribution (λ = 8)")
plt.xlabel("Number of Emergency Cases per Hour")
plt.ylabel("Probability")
plt.show()
```


### 2. For normal distribution (Mean=100, Std=15):
- Compute probability score above 130.
- Find Z-score.
- Plot PDF and CDF.


```python
mean = 100 
std = 15
score = 130 

z_score = (score - mean) / std 
prob_above_130 = 1 - norm.cdf(score, mean, std)

print("\nNormal Distribution")
print("Z-score =", round(z_score, 4))
print("Probability (Score > 130) =", round(prob_above_130, 6))

x_norm = np.linspace(40, 160, 500)
pdf_values = norm.pdf(x_norm, mean, std)

plt.figure() 
plt.plot(x_norm, pdf_values)
plt.title("Normal Distribution PDF (Mean=100, Std=15)")
plt.xlabel("Score") 
plt.ylabel("Density")
plt.show()
```

```python
cdf_values = norm.cdf(x_norm, mean, std)

plt.figure()
plt.plot(x_norm, cdf_values)
plt.title("Normal Distribution CDF (Mean=100, Std=15)")
plt.xlabel("Score")
plt.ylabel("Cumulative Probability")
plt.show() 
```


### 3. Explain difference between discrete (Poisson) and continuous (Normal) probability.


## 1. Poisson Distribution (Discrete Probability)

- Used for counting number of events.
- Values are **discrete integers** (0, 1, 2, 3, ...).
- Example: Number of hospital emergencies per hour.
- Uses **PMF (Probability Mass Function)**.
- Probability is calculated for an **exact value**.

### Formula (Poisson PMF)

P(X = k) = (e^(-λ) * λ^k) / k!

Where:
- λ = average number of events
- k = number of occurrences
- e ≈ 2.718

### Key Properties:
- Mean = λ
- Variance = λ
- Discrete distribution

---

## 2. Normal Distribution (Continuous Probability)

- Used for measuring continuous data.
- Values can be **any real number** (like 98.5, 102.7, etc.).
- Example: Exam scores, heights, weights.
- Uses **PDF (Probability Density Function)**.
- Probability is calculated over an **interval**, not exact value.

### Formula (Normal PDF)

f(x) = (1 / (σ√2π)) * e^(-(x - μ)² / (2σ²))

Where:
- μ = mean
- σ = standard deviation

### Key Properties:
- Mean = μ
- Variance = σ²
- Bell-shaped curve
- Continuous distribution

---

## 3. Comparison Table

| Feature | Poisson (Discrete) | Normal (Continuous) |
|----------|--------------------|---------------------|
| Type | Discrete | Continuous |
| Data | Counts events | Measures values |
| Function Used | PMF | PDF |
| Example | Calls per hour | Student marks |
| Exact Probability | P(X = k) | P(X = a) = 0 |
| Interval Probability | Not required | P(a < X < b) |
