# Q.1
### Write a Python program to calculate the Euclidean distance between two points in 2D space.
1. Example: Point A(2, 3), Point B(7, 8).
2. Use both manual formula and scipy.spatial.distance.euclidean.

```python
import math 
x1, y1 = 2, 3
x2, y2 = 7, 8

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print("Eucclidean Distance (Manual Formula) :",distance)
```

```python
from scipy.spatial import distance 
pointA = (2,3)
pointB = (7,8)

dist = distance.euclidean(pointA, pointB)
print("Euclidean DIstance (Scipy) : ",dist)
```

## Q.5
- Write a Python function that converts any given vector into a unit vector.
- Input: [3, 4]
- Output: normalized unit vector.

!["Unit vector"](images/Q5_1.png)

```python
def to_unit_vector(vector):
    v = np.array(vector, dtype=float)
    magnitude = np.linalg.norm(v)

    if magnitude == 0: 
        raise ValueError("Zero vector cannot be normalized.")

    unit_vector = v / magnitude
    return unit_vector

input_vector = [3,4]
result = to_unit_vector(input_vector)
print("Original Vector : ",input_vector)
print("Unit Vector : ",result)
```

## Q.6
- Generate 5 random vectors (3D) and normalize each into unit vectors. Verify that their length is always 1.

```python
vectors = np.random.randint(-10, 10, size=(5, 3))
print("Original 3D Vectors:\n", vectors)
```

```python
unit_vectors = []
lengths = []

for v in vectors: 
    norm = np.linalg.norm(v)

    if norm == 0 : 
        unit_v = v 
    else : 
        unit_v = v / norm 
    unit_vectors.append(unit_v)
    lengths.append(np.linalg.norm(unit_v))
    
unit_vectors = np.array(unit_vectors)
print("\nUnit Vectors:\n", unit_vectors)

print("\nLengths of Unit Vectors : ")
print([float(length) for length in lengths ])
```

## Q.7
- Compute the dot product of two vectors manually and using numpy.dot.
- Example: a = [1, 2, 3], b = [4, 5, 6].

!["dot product"](images/Q7.png)

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

manual_dot = 0 
for i in range(len(a)):
    manual_dot += a[i] * b[i] 
print("Dot Product : ",manual_dot)
```

```python
# Using numpy.dot
numpy_dot = np.dot(a, b)

print("NumPy Dot Product:", numpy_dot)
```

###  Dot Product – Key Concepts

### 🔹 Interpretation of Results

- If result = 0 → vectors are **perpendicular (orthogonal)**  
- If result > 0 → angle between vectors is **less than 90°**  
- If result < 0 → angle between vectors is **greater than 90°**  

### 🔹 Formula 
**Dot Product:**  a · b = Σ (aᵢ bᵢ) \
or \
**Dot Product Formula:**  a · b = Σ aᵢ bᵢ


## Q.8
- Generate two random vectors and use the dot product to check if they are orthogonal (dot product = 0).


```python
import numpy as np

v1 = np.random.randint(-5, 6, 3)
v2 = np.random.randint(-5, 6, 3)

print("Vector 1:", v1)
print("Vector 2:", v2)

dot_product = np.dot(v1, v2)
print("Dot Product:", dot_product)

if dot_product == 0:
    print("The vectors are orthogonal (perpendicular).")
else:
    print("The vectors are NOT orthogonal.")
```

###  Orthogonality Condition

If: \
v₁ · v₂ = 0
Then:
- The angle between the vectors is **90°**
- The vectors are **perpendicular**
- The vectors are **orthogonal**

## Q.9
- Compute the cross product of two 3D vectors using both manual formula and numpy.cross.
- Example: u = [2, 3, 4], v = [5, 6, 7].

!["two 3D vectors"](images/Q9.png)

```python
u = np.array([2, 3, 4])
v = np.array([5, 6, 7])

# Manual cross product
manual_cross = np.array([
    u[1]*v[2] - u[2]*v[1],
    u[2]*v[0] - u[0]*v[2],
    u[0]*v[1] - u[1]*v[0]
])

print("Manual Cross Product:", manual_cross.tolist())
```

```python
numpy_cross = np.cross(u, v)
print("NumPy Cross Product:", numpy_cross)
```

## Q.10
- Generate three 3D vectors and use the cross product to check if they form a right-handed system (determinant > 0).

```python
v1 = np.random.randint(-5, 6, 3)
v2 = np.random.randint(-5, 6, 3)
v3 = np.random.randint(-5, 6, 3)

print("Vector v1:", v1)
print("Vector v2:", v2)
print("Vector v3:", v3)
```

```python
triple_product = np.dot(v1, np.cross(v2, v3))

print("Scalar Triple Product (Determinant):", triple_product)

if triple_product > 0:
    print("The vectors form a RIGHT-HANDED system.")
elif triple_product < 0:
    print("The vectors form a LEFT-HANDED system.")
else:
    print("The vectors are coplanar (Determinant = 0).")
```
