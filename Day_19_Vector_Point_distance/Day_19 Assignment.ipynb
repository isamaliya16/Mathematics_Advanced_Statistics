# Vector Concepts ---

## 1. What is Euclidean Distance?

**Euclidean distance** is the straight-line distance between two points in space.

For two points:

A(x₁, y₁) and B(x₂, y₂)

The formula is:

d = √[(x₂ − x₁)² + (y₂ − y₁)²]

It represents the shortest path between two locations (like GPS straight-line distance).

---

## 2. Difference Between L1 and L2 Norms

### L1 Norm (Manhattan Distance)

Formula:

||V||₁ = |x| + |y|

- Adds absolute values of components  
- Measures distance like moving in city blocks  
- Used in sparse models (Lasso, feature selection)

---

### L2 Norm (Euclidean Norm)

Formula:

||V||₂ = √(x² + y²)

- Straight-line length of vector  
- Most common distance measure  
- Used in geometry, physics, and machine learning  

---

### Key Differences

| L1 Norm | L2 Norm |
|----------|----------|
| Block-style distance | Straight-line distance |
| Produces diamond shape | Produces circular shape |
| Encourages sparsity | Minimizes overall magnitude |

---

## 3. What is a Unit Vector?

A **unit vector** is a vector whose magnitude (length) is equal to 1.

Formula:

v̂ = v / ||v||

Why we use it:

- To represent direction only  
- To normalize data  
- To compare directions between vectors  

---

## 4. What Does Dot Product Measure?

The **dot product** measures the alignment or similarity between two vectors.

Formula:

v · w = |v||w| cos(θ)

Where:
- θ is the angle between vectors  

Interpretation:

- Positive → Vectors point in similar direction  
- Zero → Vectors are perpendicular  
- Negative → Vectors point in opposite direction  

Applications:
- Finding angle between vectors  
- Measuring similarity  
- Projection in physics and machine learning  

---

## 5. When Do We Use Cross Product?

The **cross product** is used only in 3D vectors.

It produces:

- A vector perpendicular to both input vectors  
- The area of the parallelogram formed by the vectors  
- Orientation (right-hand rule)

Applications:

- Physics (torque, angular momentum)  
- Computer graphics (surface normals)  
- Robotics and navigation  
- Determining orientation (right-handed or left-handed system)

```python
import numpy as np
```

```python
V1 = np.array([3, 4])
V2 = np.array([8, 10])

print("V1:", V1)
print("V2:", V2)
```

## 1. Compute Euclidean Distance

```python
distance = np.linalg.norm(V2 - V1)
print("\nEuclidean Distance between V1 and V2:", distance)
```

## 2. Compute L1 and L2 Norms

```python
L1_V1 = np.linalg.norm(V1, ord=1)
L2_V1 = np.linalg.norm(V1)

L1_V2 = np.linalg.norm(V2, ord=1)
L2_V2 = np.linalg.norm(V2)

print("\nL1 Norm of V1:", L1_V1)
print("L2 Norm of V1:", L2_V1)

print("\nL1 Norm of V2:", L1_V2)
print("L2 Norm of V2:", L2_V2)
```

## 3. Convert V1 into Unit Vector

```python
unit_V1 = V1 / np.linalg.norm(V1)
print("\nUnit Vector of V1:", unit_V1)
```

## 4. Compute Dot Product

```python
dot_product = np.dot(V1, V2)
print("\nDot Product of V1 and V2:", dot_product)
```

## 5. Compute Cosine Similarity

```python
cosine_similarity = dot_product / (np.linalg.norm(V1) * np.linalg.norm(V2))
print("Cosine Similarity:", cosine_similarity)
```

## 6. Create 3D Vectors and Compute Cross Product

```python
V1_3D = np.array([3, 4, 0])
V2_3D = np.array([8, 10, 0])

cross_product = np.cross(V1_3D, V2_3D)
print("\nCross Product (3D):", cross_product)
```

```python
# Geometric Interpretation Output
if cosine_similarity > 0.9:
    print("\nInterpretation: The vehicles are moving in nearly the same direction.")
elif cosine_similarity < 0:
    print("\nInterpretation: The vehicles are moving in opposite directions.")
else:
    print("\nInterpretation: The vehicles have moderate directional similarity.")

if np.all(cross_product == 0):
    print("Vectors are parallel (no area formed).")
else:
    print("Cross product indicates perpendicular direction and parallelogram area formed.")
```
