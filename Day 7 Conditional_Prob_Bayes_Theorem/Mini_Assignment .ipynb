# 📊 Conditional Probability – Practice Questions
### 1️⃣ Define Conditional Probability
Conditional probability is the probability of an event occurring **given that another event has already occurred**.  
It is denoted as **P(A | B)**, which means the probability of event **A** occurring when **B** is known to have occurred.

---

### 2️⃣ Why Conditional Probability is Important?
Conditional probability is important because:
- It helps in **updating probabilities** based on new information
- It is used in **decision making under uncertainty**
- It forms the foundation of **Bayes’ Theorem**
- It is widely applied in **medical diagnosis, machine learning, weather forecasting, and risk analysis**

### 4️⃣ Explain Bayes' Theorem in Your Own Words
Bayes' Theorem helps us **update the probability of an event** based on new evidence.  
It combines prior knowledge with current information to calculate a more accurate probability.

In simple terms, **Bayes' Theorem tells us how beliefs should change when we get new data**.

---

## 📌 Key Takeaways
- Conditional probability deals with **dependent information**
- Independent events are unaffected by each other
- Bayes’ Theorem is a powerful tool for **probability updating**
- These concepts are essential for **real-world problem solving**

# 📝 Mini Assignment: Bayes’ Theorem – Spam Filter Case Study

## 📊 Step 1: Small Record Dataset

Assume an email system receives **100 emails** in total.

| Email Type | Number of Emails |
|-----------|----------------|
| Spam | 20 |
| Not Spam | 80 |
| **Total** | **100** |

---

## 📊 Step 2: Detection Results

Given:
- 95% of spam emails are correctly detected
- 10% of non-spam emails are wrongly detected as spam

| Email Category | Detected as Spam | Not Detected |
|---------------|-----------------|--------------|
| Spam (20) | 19 | 1 |
| Not Spam (80) | 8 | 72 |
| **Total** | **27** | **73** |

---

## ❓ Step 3: Question Problem Statement

An email is detected as spam by the filter.

**What is the probability that the email is actually spam?**

### P(Spam∣Detected) 
---

![python](images\Spam_Detected.png)

---

### ✔ Interpretation
There is approximately a **70% probability** that an email detected as spam is actually spam.

---


```python
# Bayes' Theorem - Spam Filter Example

total_emails = 100

spam_emails = 20
not_spam_emails = 80

P_spam = spam_emails / total_emails
P_not_spam = not_spam_emails / total_emails

P_detected_given_spam = 0.95
P_detected_given_not_spam = 0.10

# Calculate total probability of detection
P_detected = (P_detected_given_spam * P_spam) + \ (P_detected_given_not_spam * P_not_spam)

# Apply Bayes' Theorem
P_spam_given_detected = (P_detected_given_spam * P_spam) / P_detected

print("Bayes' Theorem - Spam Filter Case Study")
print("--------------------------------------")
print("P(Spam) =", P_spam)
print("P(Detected | Spam) =", P_detected_given_spam)
print("P(Detected | Not Spam) =", P_detected_given_not_spam)
print()
print("P(Detected) =", round(P_detected, 4))
print("P(Spam | Detected) =", round(P_spam_given_detected, 4))
```

## 📊 Insights from the Spam Filter Case Study

- Even with a **high detection accuracy (95%)**, not all emails detected as spam are truly spam.
- A **false positive rate** (10% of non-spam emails) significantly impacts the final probability.
- Bayes’ Theorem shows that **prior probability matters** — since only 20% of emails are spam, the final result is reduced.
- The probability that a detected email is actually spam is approximately **70%**, not 95%.
- This highlights the importance of **Bayesian reasoning** in real-world systems like spam filters, medical tests, and fraud detection.

---

### 📌 Key Insight
Bayes’ Theorem helps us make **better decisions under uncertainty** by combining prior knowledge with observed evidence.
