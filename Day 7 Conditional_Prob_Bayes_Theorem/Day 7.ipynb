# Q : 1
## Conditional Probability 
- Deck : standard 52 cards. 
- Event A : first draw is a Heart.
- Event B : second draw is Heart (without replacement).
- Do : Compute P(B|A) snd P(A_and_B)

```python
# Total cards and hearts
total_cards = 52
total_hearts = 13

# Probability of A (first card is a heart)
P_A = total_hearts / total_cards

# After drawing one heart
remaining_cards = total_cards - 1
remaining_hearts = total_hearts - 1

# Conditional probability P(B|A)
P_B_given_A = remaining_hearts / remaining_cards

# Joint probability P(A and B)
P_A_and_B = P_A * P_B_given_A

# Print results properly
print("Probability of Event A (First card is a Heart):")
print("P(A) =", round(P_A, 4))

print("\nConditional Probability (Second card is a Heart | First card is a Heart):")
print("P(B|A) =", round(P_B_given_A, 4))

print("\nJoint Probability (First and Second cards are Hearts):")
print("P(A ∩ B) =", round(P_A_and_B, 4))
```

### 📌 Interpretation
- Once a Heart is drawn first, the chance of another Heart decreases
- Events A and B are dependent
- Conditional probability captures this dependency perfectly

# Q : 3 Condotional Probability 
#### Past data (100 weekdays)
- Rain & Traffic Jam : 22 days
- Rain & No Jam : 8 days
- No Rain & Traffic Jam : 18 days
- No Rain & No Jam : 52 days
- **Do** : Compute P(Jam | Rain) , P(Rain | Jam) and check independence.


![conditional_probability](images\Q3_1.png)
![conditional_probability](images\Q3_2.png)
![conditional_probability](images\Q3_3.png)
![conditional_probability](images\Q3_4.png)
![conditional_probability](images\Q3_5.png)

```python
# Given data
total_days = 100
rain_jam = 22
rain_no_jam = 8
no_rain_jam = 18
no_rain_no_jam = 52

# Totals
rain_days = rain_jam + rain_no_jam
jam_days = rain_jam + no_rain_jam

# Probabilities
P_rain = rain_days / total_days
P_jam = jam_days / total_days
P_rain_and_jam = rain_jam / total_days

# Conditional probabilities
P_jam_given_rain = rain_jam / rain_days
P_rain_given_jam = rain_jam / jam_days

# Independence check
independent = P_rain_and_jam == (P_rain * P_jam)

# Output
print("P(Jam | Rain) =", round(P_jam_given_rain, 3))
print("P(Rain | Jam) =", round(P_rain_given_jam, 3))
print("P(Rain ∩ Jam) =", P_rain_and_jam)
print("P(Rain) * P(Jam) =", round(P_rain * P_jam, 3))
print("Are Rain and Jam independent?", independent)
```

# Q : 4 Bayes' Theorem - Email Spam Filter 
- P(Spam) - 0.3 , P(Not Spam) - 0.7
- Classifier : keyword "WIN" appears with
- P(WIN | Spam) = 0.6
- P(WIN | Not Spam) = 0.5
- **DO :** Compute P(Spam | WIN)

![Bayes' Theorem](images\Q4_1.png)
![Bayes' Theorem](images\Q4_2.png)

```python
# Given probabilities
P_spam = 0.3
P_not_spam = 0.7

P_win_given_spam = 0.6
P_win_given_not_spam = 0.5

# Total probability of WIN
P_win = (P_win_given_spam * P_spam) + \
        (P_win_given_not_spam * P_not_spam)

# Bayes' Theorem
P_spam_given_win = (P_win_given_spam * P_spam) / P_win

# Output
print("P(WIN) =", round(P_win, 3))
print("P(Spam | WIN) =", round(P_spam_given_win, 3))
```

# Q : 5 Bayes' Theorem - Medical Test (Sensitivity/Specificity)
- Disease prevalence: P(D) = 0.02
- Test sensitivity : P(+ID) = 0.95
- Specificity : P(-|-D) = 0.95 ==> P(+ | -D) = 0.02
- **Do :** Compute P(D|+) and P(-D|+)
- here, Symbol "~" represents "not"

![Bayes' Theorem](images\Q5_1.png)
![Bayes' Theorem](images\Q5_2.png)
![Bayes' Theorem](images\Q5_3.png) 

```python
# Given probabilities
P_D = 0.02          # P(D)
P_not_D = 1 - P_D   # P(~D)

P_pos_given_D = 0.95     # Sensitivity: P(+ | D)
P_pos_given_not_D = 0.02  # Given: P(+ | ~D)

# Total probability of positive test
P_positive = (P_pos_given_D * P_D) + \
             (P_pos_given_not_D * P_not_D)

# Bayes' theorem calculations
P_D_given_pos = (P_pos_given_D * P_D) / P_positive
P_not_D_given_pos = (P_pos_given_not_D * P_not_D) / P_positive

# Output results
print("P(Positive Test) =", round(P_positive, 4))
print("P(D | +) =", round(P_D_given_pos, 4))
print("P(~D | +) =", round(P_not_D_given_pos, 4))
```

# Q : 6 Bayes' Theorem - Manufacturing Quality Control
- Three machines produce : M1:50, M2:30%, M3:20%
- Defect rates : P(D|M1)=0.01, P(D|M2)=0.03, P(D|M3)=0.05
- A product is randomly selected and found defective.
- **DO :** Compute P(M1|D), P(M2|D), P(M3|D)

![Bayes' Theorem](images\Q6_1.png)
![Bayes' Theorem](images\Q6_2.png)

```python
# Machine production probabilities
P_M1 = 0.50
P_M2 = 0.30
P_M3 = 0.20

# Defect probabilities
P_D_given_M1 = 0.01
P_D_given_M2 = 0.03
P_D_given_M3 = 0.05

# Total probability of a defective product
P_D = (P_D_given_M1 * P_M1) + \
      (P_D_given_M2 * P_M2) + \
      (P_D_given_M3 * P_M3)

# Bayes' theorem calculations
P_M1_given_D = (P_D_given_M1 * P_M1) / P_D
P_M2_given_D = (P_D_given_M2 * P_M2) / P_D
P_M3_given_D = (P_D_given_M3 * P_M3) / P_D

# Output results
print("P(M1 | D) =", round(P_M1_given_D, 4))
print("P(M2 | D) =", round(P_M2_given_D, 4))
print("P(M3 | D) =", round(P_M3_given_D, 4))
```
