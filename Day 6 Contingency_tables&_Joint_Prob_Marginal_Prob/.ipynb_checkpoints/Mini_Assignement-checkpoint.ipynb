# 📊 Mini Assignment  
## Student Pass/Fail vs Gender – Probability Analysis

---

## 🎯 Objective
To create a small dataset of students categorized by **Gender** and **Exam Result (Pass/Fail)** and analyze:
- Joint Probability  
- Marginal Probability  

---

## 🧾 Dataset Description

We consider **10 students** with the following attributes:
- Gender: Male / Female
- Result: Pass / Fail

| Student | Gender | Result |
|-------|--------|--------|
| 1 | Male | Pass |
| 2 | Female | Pass |
| 3 | Male | Fail |
| 4 | Female | Pass |
| 5 | Male | Pass |
| 6 | Female | Fail |
| 7 | Male | Pass |
| 8 | Female | Pass |
| 9 | Male | Fail |
|10 | Female | Pass |

---


```python
import pandas as pd 
data = {
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Result": ["Pass", "Pass", "Fail", "Pass", "Pass",
               "Fail", "Pass", "Pass", "Fail", "Pass"]
}

df = pd.DataFrame(data)

# Total observations
total = len(df)

# Joint Probabilities
p_male_pass = len(df[(df["Gender"] == "Male") & (df["Result"] == "Pass")]) / total
p_female_pass = len(df[(df["Gender"] == "Female") & (df["Result"] == "Pass")]) / total

# Marginal Probabilities
p_pass = len(df[df["Result"] == "Pass"]) / total
p_fail = len(df[df["Result"] == "Fail"]) / total

print("P(Male & Pass):", p_male_pass)
print("P(Female & Pass):", p_female_pass)
print("P(Pass):", p_pass)
print("P(Fail):", p_fail)

```
