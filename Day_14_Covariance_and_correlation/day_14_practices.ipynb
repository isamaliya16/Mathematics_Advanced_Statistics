# 🔹 Part 1: Covariance

Covariance measures how two variables move together.

It measures the direction of relationship between two variables.
- Positive → move together
- Negative → move opposite
- Near 0 → no linear relationship


image.png

```python
# Covariance Calculation

import numpy as np

# -----------------------------
# Step 1: Sample Data
# -----------------------------
X = np.array([2, 4, 6, 8, 10])     # Study Hours
Y = np.array([50, 55, 65, 70, 80]) # Exam Scores

n = len(X)

# -----------------------------
# Step 2: Manual Covariance
# -----------------------------
mean_X = X.mean()
mean_Y = Y.mean()

cov_manual = ((X - mean_X) * (Y - mean_Y)).sum() / (n - 1)

# -----------------------------
# Step 3: NumPy Covariance
# -----------------------------
cov_numpy = np.cov(X, Y)[0, 1]

# -----------------------------
# Final Output
# -----------------------------
print("Mean of X:", mean_X)
print("Mean of Y:", mean_Y)

print("\nManual Covariance:", cov_manual)
print("NumPy Covariance:", cov_numpy)

# Interpretation
if cov_manual > 0:
    relation = "Positive Relationship: As study hours increase, scores increase"
elif cov_manual < 0:
    relation = "Negative Relationship: As one increases, the other decreases"
else:
    relation = "No Linear Relationship"

print("\nRelationship:", relation)
```


# 🔹 Part 2: Pearson Correlation

Correlation measures the strength and direction of a linear relationship between two variables.

- +1 → Perfect positive relationship
- -1 → Perfect negative relationship
- 0 → No linear relationship


image.png

```python
# Correlation Calculation

import numpy as np
from scipy import stats

# -----------------------------
# Step 1: Sample Data
# -----------------------------
X = np.array([2, 4, 6, 8, 10])     # Study Hours
Y = np.array([50, 55, 65, 70, 80]) # Exam Scores

n = len(X)

# -----------------------------
# Step 2: Manual Pearson Correlation
# -----------------------------
mean_X = X.mean()
mean_Y = Y.mean()

numerator = ((X - mean_X) * (Y - mean_Y)).sum()

denominator = np.sqrt(
    ((X - mean_X)**2).sum() *
    ((Y - mean_Y)**2).sum()
)

r_manual = numerator / denominator

# -----------------------------
# Step 3: NumPy Correlation
# -----------------------------
r_numpy = np.corrcoef(X, Y)[0, 1]

# -----------------------------
# Step 4: SciPy Correlation
# -----------------------------
r_scipy, p_value = stats.pearsonr(X, Y)

# -----------------------------
# Final Output
# -----------------------------
print("Mean of X:", mean_X)
print("Mean of Y:", mean_Y)

print("\nManual Correlation (r):", r_manual)
print("NumPy Correlation:", r_numpy)
print("SciPy Correlation:", r_scipy)

# Interpretation
if r_manual > 0:
    relation = "Positive Linear Relationship"
elif r_manual < 0:
    relation = "Negative Linear Relationship"
else:
    relation = "No Linear Relationship"

print("\nRelationship:", relation)
```


# 🔹 Part 3: Spearman Rank Correlation

Spearman correlation measures the strength and direction of a monotonic relationship between two variables using ranks instead of raw values.

- Works for non-linear but monotonic relationships
- Used when data is ordinal
- Less sensitive to outliers



image.png

```python
# Spearman Rank Correlation

import numpy as np
from scipy.stats import spearmanr, rankdata

# -----------------------------
# Step 1: Sample Data
# -----------------------------
X = np.array([85, 90, 78, 92, 88])  # Academic Scores
Y = np.array([3, 1, 5, 2, 4])       # Sports Rank

n = len(X)

# -----------------------------
# Step 2: Convert to Ranks (Manual)
# -----------------------------
rank_X = rankdata(X)
rank_Y = rankdata(Y)

# -----------------------------
# Step 3: Compute Differences in Ranks
# -----------------------------
d = rank_X - rank_Y
d_squared = d ** 2

# -----------------------------
# Step 4: Manual Spearman Formula
# -----------------------------
rho_manual = 1 - (6 * d_squared.sum()) / (n * (n**2 - 1))

# -----------------------------
# Step 5: SciPy Spearman
# -----------------------------
rho_scipy, p_value = spearmanr(X, Y)

# -----------------------------
# Final Output
# -----------------------------
print("Original X:", X)
print("Original Y:", Y)

print("\nRank X:", rank_X)
print("Rank Y:", rank_Y)

print("\nManual Spearman Correlation (rho):", rho_manual)
print("SciPy Spearman Correlation:", rho_scipy)

# Interpretation
if rho_manual > 0:
    relation = "Positive Monotonic Relationship"
elif rho_manual < 0:
    relation = "Negative Monotonic Relationship"
else:
    relation = "No Monotonic Relationship"

print("\nRelationship:", relation)
```
