# Question Q.1 
### Poisson Distribution: Call Center
#### A call center receives on average 5 calls per hour.
**Find:**
- 1. Probability of receiving exactly 3 calls in an hour.
- 2. Probability of receiving more than 6 calls in an hour.
- 3. Simulate 200 hours of calls and plot histogram.

!["Poisson"](images/Q1_1.png)

!["Poisson"](images/Q1_2.png)

```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.stats import poisson 
```

```python
lambda_value = 5 
# 1.
p_exact_3 = poisson.pmf(3, lambda_value)
print("P(X=3) : ",round(p_exact_3,4))

# 2. 
p_more_6 = 1 - poisson.cdf(6, lambda_value) 
print("P(X>6) : ",round(p_more_6),4)

# 3. 
np.random.seed(42)
simulated_calls = np.random.poisson(lambda_value, 200)

plt.hist(simulated_calls, bins=range(0, max(simulated_calls)+2),edgecolor='black',alpha=0.7)
plt.title("Simulation of 200 Hours - Call Center")
plt.xlabel("Number of Calls per Hour")
plt.ylabel("Frequency")
plt.show()
```

### Interpretation
- Most hours will have around 4–6 calls.
- Getting exactly 3 calls happens about 14% of the time.
- Getting more than 6 calls happens about 24% of the time.
- Histogram will show a right-skewed discrete distribution centered around 5.

# Q.2 Poisson Distribution: Hospital Emergency Visits
### Average number of patients arriving at an emergency ward is 2 per 15 minutes.
**Find:**
- 1. Probability that exactly 4 patients arrive in 30 minutes.
- 2. Probability that fewer than 2 patients arrive in 30 minutes. 
- 3. Verify with a simulation of 500 trials. 


!["Poisson Distribution"](images/Q2_1.png)

!["Poisson Distribution"](images/Q2_2.png)

```python
import numpy as np 
from scipy.stats import poisson 
import matplotlib.pyplot as plt 
```

```python
lambda_value = 4 
p_exact_4 = poisson.pmf(4,lambda_value)
p_less_2 = poisson.cdf(1, lambda_value)

print("Theoretical P(X=4):", round(p_exact_4, 4))
print("Theoretical P(X<2):", round(p_less_2, 4))
```

```python
# Simulation of 500 trials
np.random.seed(42)
simulated_data = np.random.poisson(lambda_value, 500)

# Empirical probabilities
sim_p_exact_4 = np.mean(simulated_data == 4)
sim_p_less_2 = np.mean(simulated_data < 2)

print("Simulated P(X=4):", round(sim_p_exact_4, 4))
print("Simulated P(X<2):", round(sim_p_less_2, 4))

# Histogram
plt.hist(simulated_data, bins=range(0, max(simulated_data)+2),
         edgecolor='black', alpha=0.7)

plt.title("Simulation of 500 Trials (30 min Emergency Visits)")
plt.xlabel("Number of Patients")
plt.ylabel("Frequency")
plt.show()
```

### Interpretation
- About 19.5% of the time exactly 4 patients arrive.
- About 9% of the time fewer than 2 patients arrive.
- Simulation results should be very close to theoretical values.
- Distribution is discrete and centered around 4 patients.

# Q.3 Z-Score Probability: Exam Marks
### In a class, marks are normally distributed with mean 70, SD = 10.
**Find:**
- 1. Z-score of a student scoring 85.
- 2. Probability of scoring above 85.
- 3. Probability of scoring between 60 and 80.
- Z-Score Probability: Heights of Students

!["Z-Score"](images/Q3_1.png)


!["Z-Score"](images/Q3_2.png)

!["Z-Score"](images/Q3_3.png)

```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.stats import norm
```

```python
print("H0: μ = 70 (The student's marks are equal to the class average.)")
print("H1: μ > 70 (The student's marks are significantly greater than the class average.)")

mean = 70
sd = 10 
x = 85

z_score = (x-mean)/sd
print("Z-score of 85 : ",round(z_score,2))

prob_above_85 = 1 - norm.cdf(x, mean, sd)
print("P(X > 85) : ",round(prob_above_85, 4))

prob_between = norm.cdf(80, mean, sd) - norm.cdf(60, mean, sd)
print("P(60 < X < 80):", round(prob_between, 4))

x_values = np.linspace(30, 110, 1000)
y_values = norm.pdf(x_values, mean, sd)

plt.plot(x_values, y_values)
plt.fill_between(x_values, y_values, where=(x_values > 85), alpha=0.3)
plt.fill_between(x_values, y_values, where=((x_values > 60) & (x_values < 80)), alpha=0.3)

plt.title("Normal Distribution of Exam Marks")
plt.xlabel("Marks")
plt.ylabel("Probability Density")
plt.show()
```

### Insight
- A score of **85** is **1.5 standard deviations above the mean**, indicating strong performance.
- Only **6.68%** of students score above 85, so this mark is relatively rare and above average.
- Around **68.26%** of students score between 60 and 80, which represents the typical performance range (within ±1 standard deviation).


# Q.4
### Heights of students are normally distributed with mean 165 cm, SD = 8 cm.
**Find:**
- 1. Z-score of a student with height 180 cm.
- 2. Probability that a randomly selected student is taller than 180 cm.
- 3. Probability that height lies between 160 and 170 cm.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 165
sd = 8

z_180 = (180 - mean) / sd
prob_above_180 = 1 - norm.cdf(180, mean, sd)
prob_between = norm.cdf(170, mean, sd) - norm.cdf(160, mean, sd)

print("Z-score of 180 cm:", round(z_180, 3))
print("P(X > 180):", round(prob_above_180, 4))
print("P(160 < X < 170):", round(prob_between, 4))

x = np.linspace(140, 190, 1000)
y = norm.pdf(x, mean, sd)

plt.figure()
plt.plot(x, y, color="darkblue", linewidth=2)

plt.plot(x, y)
plt.fill_between(x, y, where=((x >= 160) & (x <= 170)),
                 color="green", alpha=0.5, label="160–170 cm")

plt.fill_between(x, y, where=(x >= 180),
                 color="red", alpha=0.5, label="Above 180 cm")


plt.axvline(mean, color="black", linestyle="--", label="Mean (165 cm)")
plt.axvline(180, color="red", linestyle="--")
plt.axvline(160, color="green", linestyle="--")
plt.axvline(170, color="green", linestyle="--")

plt.title("Normal Distribution of Student Heights", fontsize=14)
plt.xlabel("Height (cm)", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Q.5
### PDF & CDF: Normal Distribution of Weights
Weights of bags are normally distributed with mean = 50 kg, SD = 5 kg. \
**Find:**
- 1. PDF value at x = 55.
- 2. CDF value at x = 55.
- 3. Plot PDF and CDF curves for the data.

```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.stats import norm 

mean = 50 
sd = 5 

pdf_55 = norm.pdf(55, mean, sd)
cdf_55 = norm.cdf(55, mean, sd)
print("PDF at x = 55:", round(pdf_55, 5))
print("CDF at x = 55:", round(cdf_55, 5))

x = np.linspace(30, 70, 1000)
y_pdf = norm.pdf(x, mean, sd)

plt.figure()
plt.plot(x, y_pdf)
plt.axvline(55)
plt.title("PDF of Normal Distribution (Weights)")
plt.xlabel("Weight (kg)")
plt.ylabel("Probability Density")
plt.show()

y_cdf = norm.cdf(x, mean, sd)

plt.figure()
plt.plot(x, y_cdf)
plt.axvline(55)
plt.title("CDF of Normal Distribution (Weights)")
plt.xlabel("Weight (kg)")
plt.ylabel("Cumulative Probability")
plt.show()
```

# Q.6 PDF & CDF: Exponential Distribution (Service Time)
### Service times (in minutes) at a counter follow an exponential distribution with 2 = 0.2.

**Find:**
- 1. PDF value at x = 5.
- 2. CDF value at x = 5.
- 3. Probability service time > 10 minutes.
- 4. Plot PDF & CDF.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon
 
lam = 0.2  
scale = 1/lam

pdf_5 = expon.pdf(5, scale=scale)
cdf_5 = expon.cdf(5, scale=scale)
prob_gt_10 = 1 - expon.cdf(10, scale=scale)

print("PDF at x = 5:", round(pdf_5, 5))
print("CDF at x = 5:", round(cdf_5, 5))
print("P(X > 10):", round(prob_gt_10, 5))

x = np.linspace(0, 25, 1000)
y_pdf = expon.pdf(x, scale=scale)

plt.figure()
plt.plot(x, y_pdf)
plt.axvline(5)
plt.axvline(10)
plt.title("PDF of Exponential Distribution (Service Time)")
plt.xlabel("Service Time (minutes)")
plt.ylabel("Probability Density")
plt.show()

y_cdf = expon.cdf(x, scale=scale)

plt.figure()
plt.plot(x, y_cdf)
plt.axvline(5)
plt.axvline(10)
plt.title("CDF of Exponential Distribution (Service Time)")
plt.xlabel("Service Time (minutes)")
plt.ylabel("Cumulative Probability")
plt.show()
```
