# Q.1
### Create a 2×2 matrix A = [[4, 2], [1, 3]] and compute its eigenvalues and eigenvectors using NumPy.

```python
import numpy as np
```

```python
A = np.array([[4, 2], [1, 3]])
print("Matrix A")
print(A)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues : ",eigenvalues)
print("\nEigenvectors : \n", eigenvectors)
```

# Q.2
### Verify the relationship Av = λv manually for one eigenpair.

```python
A = np.array([[4, 2], [1, 3]])
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Matrix A:")
print(A)
```

```python
lambda_val = eigenvalues[0]
v = eigenvectors[:, 0]
lambda_v = lambda_val * v
print("\nEigenvalue (λ):")
print(lambda_val)

print("\nEigenvector (v):")
print(v)

print("\nλv:")
print(lambda_v)
```

```python
Av = np.dot(A, v)
```

```python
print("\nλv:")
print(lambda_v)
```

```python
print("\nVerification (Av ≈ λv):")
print(np.allclose(Av, lambda_v))
```
