# Q.2
### Verify the relationship Av = λv manually for one eigenpair.

!["img"](images/Q2_1.png)

```python
A = np.array([[4, 2], [1, 3]])
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Matrix A:")
print(A)
```

```python
lambda_val = eigenvalues[0]
v = eigenvectors[:, 0]
lambda_v = lambda_val * v
print("\nEigenvalue (λ):")
print(lambda_val)

print("\nEigenvector (v):")
print(v)

print("\nλv:")
print(lambda_v)
```

```python
Av = np.dot(A, v)
```

```python
print("\nλv:")
print(lambda_v)
```

```python
print("\nVerification (Av ≈ λv):")
print(np.allclose(Av, lambda_v))
```

# Q.4
### Perform LU decomposition of a 3×3 matrix using SciPy.

!["LU"](images/Q4_1.png)

!["LU"](images/Q4_3.png)

!["LU"](images/Q4_4.png)

```python
from scipy.linalg import lu 
A = np.array([[2,3,1],[4,7,7],[6,18,22]])
print("Matrix A:")
print(A)
```

```python
P, L, U = lu(A)
print("\nPermutation Matrix P:")
print(P)

print("\nLower Triangular Matrix L:")
print(L)

print("\nUpper Triangular Matrix U:")
print(U)
```

```python
# For SciPy LU:  A = P @ L @ U
reconstructed_A = P @ L @ U
print("\nReconstructed Matrix (P @ L @ U):")
print(np.round(reconstructed_A,2))

print("\nVerification (A ≈ P @ L @ U):")
print(np.allclose(A, reconstructed_A))
```

# Q.5
### Perform QR decomposition of the same matrix and verify A=QR.

!["QR decomposition"](images/Q5_1.png)

!["QR decomposition"](images/Q5_2.png)

!["QR decomposition"](images/Q5_3.png)

!["QR decomposition"](images/Q5_4.png)

!["QR decomposition"](images/Q5_5.png)

```python
A = np.array([[2, 3, 1],
              [4, 7, 7],
              [6, 18, 22]])

print("Matrix A:")
print(A)
```

```python
Q, R = np.linalg.qr(A)

print("\nMatrix Q (Orthogonal Matrix):")
print(np.round(Q, 2))

print("\nMatrix R (Upper Triangular Matrix):")
print(np.round(R, 2))
```

```python
QR = np.dot(Q, R)

print("\nProduct Q * R:")
print(np.round(QR, 2))

# Step 4: Check verification
print("\nVerification (A ≈ QR):", np.allclose(A, QR))
```
