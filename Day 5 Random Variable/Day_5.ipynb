# Q : 1 Discrete Random Variable - Dice Roll
- Simulate rolling a fair 6-sided dice 1000 times.
- store the result as a discrete random variable.
- plot the frequency of each variable. 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
from collections import Counter  

rolls = np.random.randint(1, 7, 1000)

freq = Counter(rolls)


plt.figure()
plt.bar(freq.keys(), freq.values())
plt.xlabel("Dice Face")
plt.ylabel("Frequency")
plt.title("Dice Roll Frequency (1000 Trials)")
plt.show()
```

# Q: 3 Probability Distribution - Coin Toss
- Simulate a 500 tosses of fair coin .
- Calculate the probabilitty of getting heads adn tails .
- Plot the probability distribution .

```python
tosses = np.random.choice(['H','T'],size=500)
```

```python
heads = np.sum(tosses == 'H')
tails = np.sum(tosses == 'T')

prob_heads = heads / 500
prob_tails = tails / 500

print("Probability of Heads:", prob_heads)
print("Probability of Tails:", prob_tails)
```

```python
outcomes = ['Heads','Tails']
probabilities = [prob_heads,prob_tails]
plt.figure()
plt.bar(outcomes,probabilities)
plt.xlabel("Outcomes")
plt.ylabel("Probability")
plt.title("Probability Distribbution of Coin Toss (500 Trials)")
plt.show()
```

# Q: 6 Mean - Exepected value of Coin Toss (Earnings Game)
- Define a game  : win $2 for heads, lose $1 for tails.
- Create the probability distribution .
- Calculate te mean (excepted earnings per toss.)


```python

# Game definition:
# Heads  -> Win $2
# Tails  -> Lose $1

# Probability distribution (fair coin)
outcomes = {
    2: 0.5,   # Earnings for Heads
   -1: 0.5    # Earnings for Tails
}

# Calculate expected earnings per toss
expected_earnings = sum(x * p for x, p in outcomes.items())

# Print the result
print("Expected Earnings per Toss: $", expected_earnings)
```

# Q: 7 Variance - Dice Roll 
- Define the probability distribution oof a fair dice.
- Calculate its variance using the formula.
- prit the result.

```python
outcomes = [1, 2, 3, 4, 5, 6]

p = 1 / 6

mean = sum(x * p for x in outcomes)

# Calculate variance using Var(X) = Σ (x - μ)^2 * P(x)
variance = sum((x - mean) ** 2 * p for x in outcomes)

print("Variance of a Fair Dice Roll:", variance)
```

# Q: 8 Variance - Game Earnings 
- Use the coin toss game
- Calculate the variance of earnings.
- interpret what a high/low variance means.

```python
# Q8: Variance – Game Earnings (Coin Toss)

# Earnings and probabilities
earnings = [2, -1]        # Heads = +2, Tails = -1
probabilities = [0.5, 0.5]

mean = sum(x * p for x,p in zip(earnings,probabilities))

variance = sum((x - mean) ** 2 * p for x,p in zip(earnings,probabilities))
print("Expected Earnings per Toss : ",mean)
print("Variance  of Earnings : ",variance)
```
