# Step 1: Generate Website Visits Data (Random Variable)
- We consider website visits per day as a discrete random variable 𝑋.
- Simulated data for 50 days, with values between 80 and 160 visits per day.
- Python generates this data randomly to mimic real-world fluctuations.

```python
import numpy as np
import pandas as pd

# Simulate 50 days of website visits (discrete random variable)
np.random.seed(42)
visits_data = np.random.randint(80, 161, size=31)

# Create a DataFrame
df = pd.DataFrame(visits_data, columns=['Visits'])
# df.head()
df
```

# Step 2: Create Probability Distribution
- A probability distribution shows how likely each visit count is to occur.
- Calculated as: 
P(X=x) = Frequency of x / Total number of days
