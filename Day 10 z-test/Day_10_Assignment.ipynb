```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
import math
```

```python
from scipy.stats import norm 
import math 

mu = 24
x_bar = 23.4
sigma = 2
n = 36
alpha = 0.05

print("H0: μ = 24 hours")
print("H1: μ ≠ 24 hours (two-tailed test)")

SE = sigma / math.sqrt(n)
print(f"\n Standard Error (SE) : {SE:.4f}")

z_score = (x_bar - mu) / SE
print(f"Calculated z-score : {z_score:.4f}")

z_critical = norm.ppf(1 - alpha/2) # two tails 
print(f"Critical z-value : {z_critical:.4f}")

if z_score < -z_critical or z_score > z_critical : 
    print("Reject H0 ")
    print(" interpretation = There is enough evidence to conclude that the average battery backup is different from 24 hours.")
else :
    print("We Fail to Reject H0")
    print("There is not enough evidence to conclude that the average battery backup is different from 24 hours.")

    
```

```python
z = np.linspace(-4,4,1000)
pdf = norm.pdf(z,0,1)

plt.figure(figsize=(10,6))
sns.set_style('whitegrid')
plt.plot(z,pdf,color='blue',label='Standard Normal Distribution')

plt.fill_between(z,0,pdf,where=(z <= -z_critical), alpha=0.3,label='Rejection Region (Left)')
plt.fill_between(z,0,pdf,where=(z > z_critical),alpha=0.3,label='Rejection Region (Right)')

plt.axvline(z_score, color='green',linestyle='--',linewidth=2,label=f'Calculated z = {z_score:.2f}')

plt.axvline(z_critical,linestyle='--',linewidth=1)
plt.axvline(-z_critical,linestyle='--',linewidth=1)
plt.title('Two-Tailed Z-Test Visualization', fontsize=16)
plt.xlabel('Z-Score')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```

```python
from scipy.stats import norm
import math

mu = 4              
x_bar = 4.3         
sigma = 0.7         
n = 49              
alpha = 0.05        

print("H0: μ ≤ 4 hours (claim is true)")
print("H1: μ > 4 hours (users study more than claimed)")

SE = sigma / math.sqrt(n)
print(f"\nStandard Error (SE) = {SE:.4f}")

# Calculate z-score
z_score = (x_bar - mu) / SE
print(f"Calculated z-score = {z_score:.4f}")

# Compute one-tailed p-value (right-tailed)
p_value = 1 - norm.cdf(z_score)
print(f"One-tailed p-value = {p_value:.4f}")

# Identify critical z-value
z_critical = norm.ppf(1 - alpha)  # right-tailed
print(f"Critical z-value = {z_critical:.4f}")

# Using p-value
if p_value < alpha:
    print("\nReject H0")
else:
    print("\nFail to reject H0")

# Using critical z-value
if z_score > z_critical:
    print("Reject H0")
else:
    print("Fail to reject H0")

if p_value < alpha:
    interpretation = "Users are studying more than the platform claims, showing higher engagement."
else : 
    interpretation = "No clear evidence that users study more than claimed, so engagement is as expected."

print(f"\n Final Interpretation: {interpretation}")
```

```python
Q : 3 
from scipy.stats import norm
import math

mean_a = 42
std_a = 6
n_a = 64

mean_b = 39
std_b = 6
n_b = 81

alpha = 0.05
print("H0: μA = μB (No difference in average delivery times)")
print("H1: μA ≠ μB (Average delivery times are different)")

SE = math.sqrt((std_a**2)/n_a + (std_b**2)/n_b)
print(f"\n Standard Error (SE) : {SE:.4f} ")

z_score = (mean_a - mean_b) / SE 
print(f"Calculated z-score : {z_score:.4f}")

p_value = 2 * (1 - norm.cdf(abs(z_score)))
print(f"Two - tailed p_value : {p_value}")

z_critical = norm.ppf(1 - alpha/2)
print(f"Critical z-value : ±{z_critical:.4f}")

# Method 1: Using p-value
if p_value < alpha:
    print("\nMethod 1 (p-value): Reject H0")
else:
    print("\nMethod 1 (p-value): Fail to reject H0")
    
# Method 2: Using critical z-value
if z_score > z_critical or z_score < -z_critical:
    print("Method 2 (critical value): Reject H0")
    print("conclusion = The two companies have significantly different average delivery times.")

else:
    print("Method 2 (critical value): Fail to reject H0")
    print("conclusion = There is no significant difference in average delivery times between the two companies.")
```

# Q : 4

```python
from scipy.stats import norm
import math

mean_old = 8.5
std_old = 1.5
n_old = 100

mean_new = 7.9
std_new = 1.5
n_new = 90

alpha = 0.05

print("H0: μ_new ≥ μ_old  (new algorithm is NOT faster)")
print("H1: μ_new < μ_old  (new algorithm is faster)\n")

SE = math.sqrt((std_old**2)/n_old + (std_new**2)/n_new)
print(f"Standard Error (SE) = {SE:.4f}")

z_score = (mean_new - mean_old) / SE
print(f"Calculated z-score = {z_score:.4f}")

p_value = norm.cdf(z_score)
print(f"One-tailed p-value = {p_value:.4f}")

z_critical = norm.ppf(alpha)
print(f"Critical z-value = {z_critical:.4f}")

# Method 1 — p-value
if p_value < alpha:
    print("\nMethod 1 (p-value): Reject H0")
else:
    print("\nMethod 1 (p-value): Fail to Reject H0")

# Method 2 — critical value
if z_score < z_critical:
    print("Method 2 (critical value): Reject H0")
else:
    print("Method 2 (critical value): Fail to Reject H0")

if p_value < alpha:
    tech = "Statistically significant evidence shows the new algorithm reduces response time."
    simple = "The new software is faster than the old one."
else:
    tech = "No statistically significant evidence that the new algorithm is faster."
    simple = "We cannot confirm the new software is faster."

print("\nTechnical Conclusion:")
print(tech)

print("\nNon-Technical Conclusion:")
print(simple)
```
