# Q.1  write a python code. 

- Generate a population dataset of 10,000 heights (in cm).
- Take a random sample of 100 values.
- Compare sample mean with population mean.
- Plot both to observe the difference visually using histogram. Plot vertical line for both means.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

population_size = 10000
population = np.random.normal(loc=170, scale=10, size=population_size)

sample = np.random.choice(population, size=100, replace=False)

population_mean = np.mean(population)
sample_mean = np.mean(sample)

plt.figure()
plt.hist(population, bins=50, alpha=0.7, label="Population")
plt.hist(sample, bins=20, alpha=0.7, label="Sample")

plt.axvline(population_mean, linestyle='--',
            label=f"Population Mean = {population_mean:.2f}")
plt.axvline(sample_mean, linestyle=':',
            label=f"Sample Mean = {sample_mean:.2f}")

plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.title("Comparison of Population Mean and Sample Mean")
plt.legend()

plt.show()
```

# Q.2
- Create a dataset of students' test scores.
- Assume an average passing score (like 60 marks).
- Display how students scored compared to the passing mark.
- Mention how you might ask a question like: "Are students scoring around or below 60?"


```python
np.random.seed(10)

num_students = 200
scores = np.random.normal(loc=62, scale=10, size=num_students)
passing_score = 60
average_score = np.mean(scores)
plt.figure()
plt.hist(scores,bins=20, alpha=0.8, label="Student Scores")
plt.axvline(passing_score, linestyle='--',
            label=f"Passing Mark = {passing_score}")
plt.axvline(average_score, linestyle=':',
            label=f"Average Score = {average_score:.2f}")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Students' Test Scores Compared to Passing Mark")
plt.legend()

plt.show()
```

### 📊 Observation
- Most student scores are clustered around the passing mark (60)
- The average score ≈ 62.7, which is slightly above the passing score
- Some students scored below, while many scored above the passing mark

# Q.3

- Create a dataset of 5000 daily steps walked by users.
- Randomly select a sample of 50 users.
- Compare their average steps with the population average.
- Plot both to show how closely sample represents the whole.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(25)

population_steps = np.random.normal(loc=8000, scale=2000, size=5000)

sample_steps = np.random.choice(population_steps, size=50, replace=False)

population_mean = np.mean(population_steps)
sample_mean = np.mean(sample_steps)

plt.figure()
plt.hist(population_steps, bins=50, alpha=0.7, label="Population")
plt.hist(sample_steps, bins=15, alpha=0.8, label="Sample")

plt.axvline(population_mean, linestyle='--',
            label=f"Population Avg = {population_mean:.0f}")
plt.axvline(sample_mean, linestyle=':',
            label=f"Sample Avg = {sample_mean:.0f}")

plt.xlabel("Daily Steps")
plt.ylabel("Number of Users")
plt.title("Population vs Sample Daily Steps")
plt.legend()
plt.show()
```

### 📊 Observation

- The sample average daily steps are very close to the population average. This shows that a randomly selected sample of 50 users represents the overall population of 5000 users quite well.

# Q.4
- Simulate test scores for Group A and Group B.
- Plot both distributions side by side.
- Observe and note which group performs better.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(30)

group_A = np.random.normal(loc=65, scale=8, size=200)
group_B = np.random.normal(loc=72, scale=7, size=200)

mean_A = np.mean(group_A)
mean_B = np.mean(group_B)

plt.figure()
plt.hist(group_A, bins=20, alpha=0.7, label=f"Group A (Avg = {mean_A:.2f})")
plt.hist(group_B, bins=20, alpha=0.7, label=f"Group B (Avg = {mean_B:.2f})")

plt.xlabel("Test Scores")
plt.ylabel("Number of Students")
plt.title("Test Score Distribution: Group A vs Group B")
plt.legend()
plt.show()
```

### 📊 Observation

- The distribution of Group B is shifted toward higher scores and its average score is greater than that of Group A. Therefore, Group B performs better overall than Group A.

### Q.5

- Define a business assumption like "Ice cream sales average ₹5000 per day."
- Write down null and alternate hypotheses.
- Use Python to print them clearly.


```python
assumption = "Ice cream sales average ₹5000 per day."

null_hypothesis = "H₀: Average daily ice cream sales = ₹5000"
alternate_hypothesis = "H₁: Average daily ice cream sales ≠ ₹5000"

print("Business Assumption:")
print(assumption)
print("\nHypotheses:")
print(null_hypothesis)
print(alternate_hypothesis)
```
