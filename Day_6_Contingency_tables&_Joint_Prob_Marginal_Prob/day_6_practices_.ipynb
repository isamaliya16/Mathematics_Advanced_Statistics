## 📊 Real-World Dataset: Customer Purchase Behavior


```python

import pandas as pd
import numpy as np

data = {
    "Gender": np.random.choice(["Male", "Female"], size=100),
    "Purchased": np.random.choice(["Yes", "No"], size=100, p=[0.6, 0.4])
}

df = pd.DataFrame(data)
df
```


## 1️⃣ Marginal Probability
Probability that **single event occurs**.

Screenshot 2026-02-04 at 2.46.42 PM.png


```python
p_male = len(df[df['Gender'] == 'Male']) / len(df)
p_male
```

```python
p_female = len(df[df['Gender'] == 'Female']) / len(df)
p_female
```

```python
# Marginal probability of Gender
gender_prob = df['Gender'].value_counts(normalize=True)
gender_prob
```

```python
# Marginal probability of Purchased
purchase_prob = df['Purchased'].value_counts(normalize=True)
purchase_prob
```


## 2️⃣ Joint Probability
Probability that **two events occur together**.

Screenshot 2026-02-04 at 2.50.15 PM.png


```python
p_female_yes = len(df[(df['Gender'] == 'Female') & (df['Purchased'] == 'Yes')]) / len(df)
p_female_yes
```

```python
p_male_no = len(df[(df['Gender'] == 'Male') & (df['Purchased'] == 'No')]) / len(df)
p_male_no
```


## 3️⃣ Contingency Table
A contingency table shows the frequency (counts) of combinations of two categorical variables.

- Rows → one categorical variable
- Columns → another categorical variable
- Values → counts (not probabilities)



```python

contingency_table = pd.crosstab(df["Gender"], df["Purchased"])
contingency_table
```

```python

# Joint probability table
joint_probability_table = contingency_table / contingency_table.values.sum()
joint_probability_table
```

```python

contingency_table = pd.crosstab(df["Gender"], df["Purchased"], margins=True)
contingency_table
```


## 4️⃣ Venn Diagram (Conceptual Visualization)
Note: Venn diagram represents overlapping probabilities.


```python

# Simple numeric representation for Venn understanding
female = len(df[df["Gender"] == "Female"])
purchased = len(df[df["Purchased"] == "Yes"])
female_and_purchased = len(df[(df["Gender"] == "Female") & (df["Purchased"] == "Yes")])

female, purchased, female_and_purchased
```

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

female = len(df[df["Gender"] == "Female"])
purchased = len(df[df["Purchased"] == "Yes"])
female_and_purchased = len(df[(df["Gender"] == "Female") & (df["Purchased"] == "Yes")])

purchased_only = purchased - female_and_purchased
female_only = female - female_and_purchased

venn2(
    subsets=(purchased_only, female_only, female_and_purchased),
    set_labels=('Purchased = Yes', 'Gender = Female')
)

plt.title("Venn Diagram: Purchased vs Gender")
plt.show()
```
