# Q : 1
## write a python program using matplotlib_venn to create a Venn diagram 
showing : 
- 40 students like tea
- 50 student like coffee
- 20 student like both tea and coffee 

```python
# pip install matplotlib-venn

import matplotlib.pyplot as plt 
from matplotlib_venn import venn2 

tea = 40
coffee = 50
both = 20

only_tea = tea - both
only_coffee = coffee - both 

# Total students
total_students = only_tea + only_coffee + both

# Probabilities
p_only_tea = only_tea / total_students
p_only_coffee = only_coffee / total_students
p_both = both / total_students

venn2(subsets=(only_tea, only_coffee, both),
     set_labels=('Tea Lovers','Coffee Lovers'))

plt.title("Venn Diagram of students Who Like Tea and Coffee")
plt.show()


print("Probabilities:")
print(f"P(Only Tea) = {p_only_tea:.2f}")
print(f"P(Only Coffee) = {p_only_coffee:.2f}")
print(f"P(Both Tea and Coffee) = {p_both:.2f}")
```

### 🧠 What this diagram shows
- Left circle → Students who like only Tea
- Right circle → Students who like only Coffee
- Overlapping area → Students who like both

# Q : 2 
### write a python program using matplotlib_venn to create a venn diagram for : 
- 60 people use Facebook
- 45 people use instagram
- 25  people use both platforms.
- find and display the probabilities for each region .

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

# Given data
facebook = 60
instagram = 45
both = 25

# Calculate exclusive regions
only_facebook = facebook - both
only_instagram = instagram - both
total_people = only_facebook + only_instagram + both

# Calculate probabilities
p_only_facebook = only_facebook / total_people
p_only_instagram = only_instagram / total_people
p_both = both / total_people

# Create Venn diagram
venn2(
    subsets=(only_facebook, only_instagram, both),
    set_labels=('Facebook', 'Instagram')
)

plt.title("Venn Diagram of Social Media Users")
plt.show()

# Display probabilities
print("Probabilities:")
print(f"P(Only Facebook) = {p_only_facebook:.2f}")
print(f"P(Only Instagram) = {p_only_instagram:.2f}")
print(f"P(Both Platforms) = {p_both:.2f}")
```

### 📈 Output Probabilities Explained

| Region            | Count | Probability        |
|------------------|-------|--------------------|
| Only Facebook    | 35    | 35 / 80 = **0.44** |
| Only Instagram   | 20    | 20 / 80 = **0.25** |
| Both Platforms   | 25    | 25 / 80 = **0.31** |

---

### 🧠 Statistical Meaning

- **Marginal Probability**  
  → Probability of using **Facebook** or **Instagram** individually.

- **Joint Probability**  
  → Probability of using **both platforms together**.

- **Venn Diagram Representation**  
  → Visually demonstrates **union** and **intersection** of events.

# Q : 5
### Using a contingency table of dice rolls (two dice) . write python code to calculate the marginal probabilitiy of each possible sum outcomes  .

```python
dice1 = range(1, 7)
dice2 = range(1, 7)

data = [(d1, d2, d1 + d2) for d1 in dice1 for d2 in dice2]

df = pd.DataFrame(data, columns = ["DIe1","Die1","Sum"])
```

```python
sum_table = pd.crosstab(df["Sum"],columns="Frequency")
total_outcomes = sum_table["Frequency"].sum()
sum_table["Marginal_Probability"] = sum_table['Frequency'] / total_outcomes

print("Contingency Table with Marginal Probabilities : \n ", sum_table)
```

# Q :  6
### Given a contingency table of Students study habits vs exam results, write a python code to calculate : Marginal Probability of "Pass" , Marginal Probabilities of "Fail" 

```python
data = {
    "Pass": [30, 40],
    "Fail": [10, 20]
}

index = ["Regular Study", "Irregular Study"]

contingency_table = pd.DataFrame(data, index=index)

print("Contingency Table:")
print(contingency_table)

total_students = contingency_table.values.sum()

marginal_pass = contingency_table["Pass"].sum() / total_students 
marginal_fail = contingency_table["Fail"].sum() / total_students 

print("\nMarginal Probabilities:")
print(f"P(Pass) = {marginal_pass:.2f}")
print(f"P(Fail) = {marginal_fail:.2f}")
```

### Work : 
- contingency_table["Pass"].sum() → total students who passe
- contingency_table["Fail"].sum() → total students who failed
- Divide by grand total → marginal probability

### 📈 Interpretation
- Marginal Probability of Pass → Overall chance a student passes (ignores study habit)
- Marginal Probability of Fail → Overall chance a student fails
- Marginal probabilities describe overall outcomes without conditioning on another variable.

# Q : 7 
### Given a dataset of two coin tosses , Write Python code to calculate the joint probability of getting (Head,Head), (Head,Tail), (Tail,Head) and (Tail,Tail) 

```python
data = {
    "Coin1": ["H", "H", "T", "T"],
    "Coin2": ["H", "T", "H", "T"]
}

df = pd.DataFrame(data)

print("Dataset of Two Coin Tosses:")
print(df)
```

```python
joint_table = pd.crosstab(df['Coin1'],df['Coin2'])
total_outcomes = joint_table.values.sum()

joint_probability = joint_table / total_outcomes
print("\n Joint Probability Table :")
print(joint_probability)
```

## 📊 Joint Probabilities Explained

- **P(Head, Head)** = 1 / 4 = **0.25**
- **P(Head, Tail)** = 1 / 4 = **0.25**
- **P(Tail, Head)** = 1 / 4 = **0.25**
- **P(Tail, Tail)** = 1 / 4 = **0.25**

---

## 🧠 Statistical Interpretation

- This represents a **joint probability distribution** of two coin tosses.
- Both coin tosses are **independent events**.
- All possible outcomes are **equally likely**.


# Q : 8 
### Given a dataset of fruit preference by gender , write python code to calculate the joint probability of : Male & likes Mango , Female & likes Apple .

```python
data = {
    "Gender": [
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female"
    ],
    "Fruit": [
        "Mango", "Apple", "Apple", "Mango", "Mango",
        "Apple", "Banana", "Apple", "Mango", "Banana"
    ]
}

df = pd.DataFrame(data)
```

```python
total_people = len(df)

p_male_mango = len(df[(df["Gender"] == "Male") & (df["Fruit"] == "Mango")]) / total_people
p_female_apple = len(df[(df["Gender"] == "Female") & (df["Fruit"] == "Apple")]) / total_people

print("\nJoint Probabilities:")
print(f"P(Male & likes Mango) = {p_male_mango:.2f}")
print(f"P(Female & likes Apple) = {p_female_apple:.2f}")
```

### How It Works

- Filters the dataset using **logical conditions**
- Counts the **matching rows**
- Divides by the **total number of observations**
- Result = **Joint Probability**

---

### Statistical Meaning

- **P(Male & Mango)**  
  → Probability that a randomly selected person is **male** and **prefers mango**

- **P(Female & Apple)**  
  → Probability that a randomly selected person is **female** and **prefers apple**
