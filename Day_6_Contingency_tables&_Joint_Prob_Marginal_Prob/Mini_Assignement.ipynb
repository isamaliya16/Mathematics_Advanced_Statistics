# Joint & Marginal Probability

## 1.Define Joint Probability

Joint probability is the probability of two or more events happening at the same time.

It is written as:

P(A and B)  or  P(A ∩ B)

Example:  
If  
P(A) = Probability of buying product  
P(B) = Probability of paying by card  

Then  
P(A and B) = Probability that a customer buys the product AND pays by card.

- It measures combined occurrence of events.

---

## 2.What is Marginal Probability?

Marginal probability is the probability of a single event happening, without considering other events.

It is obtained by summing or adding joint probabilities across rows or columns in a table.

It is written as:

P(A)

Example:
If we want probability of buying a product regardless of payment method,  
that is marginal probability.

- It focuses on one event only.

---

## 3.Why Contingency Tables Are Useful?

A contingency table is a table that shows the relationship between two categorical variables.

They are useful because:

- Organize joint and marginal probabilities clearly
- Show relationship between variables
- Help in hypothesis testing (Chi-square test)
- Easy comparison between categories
- Useful in business data analysis

Example:
Relationship between Gender and Product Preference.

- Contingency tables make complex data easy to understand.

# 📊 Mini Assignment  
## Student Pass/Fail vs Gender – Probability Analysis

---

## 🎯 Objective
To create a small dataset of students categorized by **Gender** and **Exam Result (Pass/Fail)** and analyze:
- Joint Probability  
- Marginal Probability  

---

## 🧾 Dataset Description

We consider **10 students** with the following attributes:
- Gender: Male / Female
- Result: Pass / Fail

| Student | Gender | Result |
|-------|--------|--------|
| 1 | Male | Pass |
| 2 | Female | Pass |
| 3 | Male | Fail |
| 4 | Female | Pass |
| 5 | Male | Pass |
| 6 | Female | Fail |
| 7 | Male | Pass |
| 8 | Female | Pass |
| 9 | Male | Fail |
|10 | Female | Pass |

---


```python
import pandas as pd 
data = {
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Result": ["Pass", "Pass", "Fail", "Pass", "Pass",
               "Fail", "Pass", "Pass", "Fail", "Pass"]
}

df = pd.DataFrame(data)

# Total observations
total = len(df)

# Joint Probabilities
p_male_pass = len(df[(df["Gender"] == "Male") & (df["Result"] == "Pass")]) / total
p_female_pass = len(df[(df["Gender"] == "Female") & (df["Result"] == "Pass")]) / total

# Marginal Probabilities
p_pass = len(df[df["Result"] == "Pass"]) / total
p_fail = len(df[df["Result"] == "Fail"]) / total

print("P(Male & Pass):", p_male_pass)
print("P(Female & Pass):", p_female_pass)
print("P(Pass):", p_pass)
print("P(Fail):", p_fail)

```
