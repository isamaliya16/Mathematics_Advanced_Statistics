# 🔹 Part 1: Eigenvalues & Eigenvectors

- An **eigenvector** of a matrix is a vector whose direction does not change after transformation by the matrix.
- It may only stretch or shrink.

- The **eigenvalue** tells us how much the eigenvector is scaled.

---

For a square matrix A and vector v:

**A v = λ v**

Where:
- A = transformation matrix
- v = eigenvector
- λ = eigenvalue

---

**Interpretation**

After transformation:
- Direction stays same
- Only magnitude changes

Example:

If: λ = 3, then Vector becomes **3 times larger**.


## 📘 Real-World Example: Feature Transformation Matrix


```python
# Eigenvalues & Eigenvectors

import numpy as np

# Define matrix
A = np.array([
    [2, 1],
    [1, 2]
])

# -----------------------------
# Step 1: Eigenvalues & Eigenvectors
# -----------------------------
eigenvalues, eigenvectors = np.linalg.eig(A)

# -----------------------------
# Step 2: Verify Av = λv
# -----------------------------
v = eigenvectors[:,0]
lambda_val = eigenvalues[0]

Av = A @ v
lambda_v = lambda_val * v

# -----------------------------
# Output
# -----------------------------
print("Matrix A:\n", A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

print("\nVerification (Av):", Av)
print("λv:", lambda_v)
```

## LU Decomposition

LU Decomposition factorizes a matrix into two matrices:

**A = L * U**

Where:
- L = Lower triangular matrix
- U = Upper triangular matrix

image.png

```python
# LU Decomposition

import numpy as np
from scipy.linalg import lu

# Define matrix
A = np.array([
    [2, 3],
    [4, 7]
])

# LU decomposition
P, L, U = lu(A)

# Verification
reconstructed_A = L @ U

print("Matrix A:\n", A)

print("\nLower Triangular Matrix L:\n", L)

print("\nUpper Triangular Matrix U:\n", U)

print("\nReconstructed A (L @ U):\n", reconstructed_A)
```

```python
import numpy as np
from scipy.linalg import lu

# Define matrix A
A = np.array([
    [2, 3],
    [4, 7]
])

# Define vector B
B = np.array([
    [5],
    [11]
])

# LU decomposition
P, L, U = lu(A)

print("Permutation Matrix P:\n", P)
print("\nLower Triangular Matrix L:\n", L)
print("\nUpper Triangular Matrix U:\n", U)

# Step 1: Compute PB
PB = P @ B

print("\nPB:\n", PB)

# Step 2: Solve Ly = PB (Forward substitution)
y = np.linalg.solve(L, PB)

print("\ny:\n", y)

# Step 3: Solve Ux = y (Backward substitution)
x = np.linalg.solve(U, y)

print("\nSolution x:\n", x)
```

### How LU Decomposition Solves Linear Systems (Ax = b)

Given a system of linear equations in matrix form **Ax = b**, where:
*   **A** is the coefficient matrix
*   **x** is the vector of unknowns
*   **b** is the constant vector

First, we perform LU decomposition on **A** such that **A = LU**. Substituting this into the original equation, we get:

**LUx = b**

To solve for **x**, we break this into two simpler systems:

1.  **Ly = b (Forward Substitution)**
    *   We introduce an intermediate vector **y** such that **Ux = y**.
    *   Since **L** is a lower triangular matrix, this system can be easily solved for **y** by starting from the first equation and substituting the values of **y** into subsequent equations.

2.  **Ux = y (Backward Substitution)**
    *   Once **y** is found, we solve for **x** using the equation **Ux = y**.
    *   Since **U** is an upper triangular matrix, this system can be easily solved for **x** by starting from the last equation and substituting the values of **x** into previous equations.

This two-step process is much more efficient than calculating the inverse of A (A<sup>-1</sup>) and then computing x = A<sup>-1</sup>b, especially for large matrices or when solving multiple systems with the same A but different b vectors.
