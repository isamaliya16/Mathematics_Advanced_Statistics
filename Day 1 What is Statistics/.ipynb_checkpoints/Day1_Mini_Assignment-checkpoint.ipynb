
## 📌 Mini Assignment

### Dataset:
Create a dataset of **10 students** with:
- Roll No
- Marks
- Gender
- City

### Tasks:
1. Identify data types.
2. Calculate average marks.
3. Create:
   - Bar chart (Gender)
   - Histogram (Marks)
4. Take a random sample of 5 students and compare average marks with population.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create dataset
data = {
    "Roll No": range(1, 11),
    "Marks": [78, 85, 92, 67, 74, 88, 90, 69, 81, 76],
    "Gender": ["Male", "Female", "Female", "Male", "Male", "Female", "Male", "Female", "Male", "Female"],
    "City": ["Ahmedabad", "Surat", "Rajkot", "Vadodara", "Ahmedabad", "Surat", "Rajkot", "Vadodara", "Ahmedabad", "Surat"]
}

df = pd.DataFrame(data)
df
```

### Data Types

```python
df.dtypes
```

### Average Marks

```python

avg_marks = df["Marks"].mean()
avg_marks
```

### Bar Chart: Gender Distribution

```python

df["Gender"].value_counts().plot(kind="bar")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Gender Distribution")
plt.show()
```

#### Pattern observed:
- The chart shows an equal distribution of male and female students, indicating balanced gender representation in the dataset.

### Histogram: Marks Distribution

```python

plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Histogram")
plt.show()
```

#### Pattern observed:
- Most students scored between 70 and 90 marks, with the highest concentration around the 80s, showing generally good academic performance and no extreme low or high outliers.

### Random Sample vs Population Average

```python

sample = df.sample(5, random_state=1)
sample_avg = sample["Marks"].mean()

population_avg = df["Marks"].mean()

print(sample)
print("Sample Average:",sample_avg)
print("Population Average:",population_avg)
```
