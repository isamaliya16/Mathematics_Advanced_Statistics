## 📝 Practice Questions (Simple Explanation with Examples)

### 1. Explain p-value in simple words
The **p-value** tells us how likely it is to get our result if the null hypothesis is true.  
**Example:** If a company claims delivery time is 30 minutes and the p-value is very small, it means the result we got is unlikely if the claim were true.

---

### 2. Difference between Confidence Interval (CI) and p-value
- A **confidence interval** gives a **range of values** where the true average may lie.  
- A **p-value** tells us **whether to accept or reject** a claim.

**Example:**  
A 95% CI of (28, 32) minutes means delivery time is likely between 28 and 32 minutes.  
A small p-value suggests the average is not 30 minutes.

---

### 3. Explain Type I & Type II Errors
- **Type I Error:** Rejecting a true null hypothesis.  
  **Example:** Saying a person is sick when they are actually healthy.

- **Type II Error:** Not rejecting a false null hypothesis.  
  **Example:** Saying a person is healthy when they are actually sick.


# 📌 Mini Assignment
### Claim: Average service time is 5 minutes.

- Assume sample data
- Calculate p-value and CI
- Identify possible errors

## 📌 Claim
Average service time = **5 minutes**

---

## 🧪 Assumed Sample Data
- Sample size (n) = **40 customers**
- Sample mean = **5.6 minutes**
- Sample standard deviation = **1.2 minutes**
- Significance level (α) = **0.05**

---

## 🧠 Hypotheses
- **Null Hypothesis (H₀):** μ = 5 minutes  
- **Alternative Hypothesis (H₁):** μ ≠ 5 minutes  

---

## 📊 Calculation (Conceptual Results)

### 1️⃣ p-value
Using a **one-sample t-test**, suppose we obtain:

- **p-value = 0.02**

**Interpretation:**  
Since **p-value < 0.05**, the result is statistically significant.

➡️ **Reject the null hypothesis**

---

### 2️⃣ 95% Confidence Interval
Assume the calculated **95% confidence interval** is:

- **CI = (5.2 , 6.0) minutes**

**Interpretation:**
- The claimed value **5 minutes** does **not lie inside** the interval.
- This supports the conclusion that the average service time is **different from 5 minutes**.

---

## ✅ Conclusion
Based on the sample data:
- The p-value indicates strong evidence against the claim.
- The confidence interval does not include **5 minutes**.

👉 **There is sufficient evidence to say the average service time is not 5 minutes.**

---

## ⚠️ Possible Errors

### 🔴 Type I Error
Rejecting **H₀** when it is actually true.  
**Example:** Concluding service time is not 5 minutes when it actually is.

### 🔵 Type II Error
Failing to reject **H₀** when it is false.  
**Example:** Concluding service time is 5 minutes when it is actually different.


```python
import numpy as np
from scipy import stats

# Claim
claimed_mean = 5

np.random.seed(0)
sample_data = np.random.normal(loc=5.6, scale=1.2, size=40)

sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
n = len(sample_data)

# One-sample t-test
t_stat, p_value = stats.ttest_1samp(sample_data, claimed_mean)

# 95% Confidence Interval
alpha = 0.05
t_critical = stats.t.ppf(1 - alpha/2, df=n-1)
margin_error = t_critical * (sample_std / np.sqrt(n))

ci_lower = sample_mean - margin_error
ci_upper = sample_mean + margin_error

print("Claimed Mean Service Time:", claimed_mean, "minutes")
print("Sample Mean:", round(sample_mean, 2))
print("Sample Standard Deviation:", round(sample_std, 2))
print("p-value:", round(p_value, 4))
print("95% Confidence Interval: (", round(ci_lower, 2), ",", round(ci_upper, 2), ")")

if p_value < alpha:
    print("Decision: Reject the null hypothesis")
else:
    print("Decision: Fail to reject the null hypothesis")
```
