# Q : 3 Pandas Dataframe & Data Types

```python
import pandas as pd

# Creating DataFrame
data = {
    "Name": ["Amit", "Neha", "Raj", "Priya"], 
    "Age": [28, 32, 30, 26],
    "Department": ["Sales", "Marketing", "Sales", "HR"],
    "Salary": [45000, 52000, 48000, 30000]
}

df = pd.DataFrame(data)
df
```

#### Check Data Types

```python
df.dtypes
```

#### Filter Numerical Data 

```python
df.select_dtypes(include=['int64','float64'])
```

#### filter Categorical Data 

```python
df.select_dtypes(include=['object'])
```

# Q : 4 Data Visulization 

```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

#### Histogram - Age Distribution 

```python
plt.hist(df['Age'],bins=5)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### Observed:
- The ages are concentrated between 26 and 32, showing a fairly compact and balanced distribution with no extreme values.

#### bar chart - Department Count

```python
df['Department'].value_counts().plot(kind='bar')
plt.title("Department Count")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

### Observed:
- The Sales department has the highest number of employees, while Marketing and HR have fewer employees.

#### pie Chart - Department Percentage

```python
df['Department'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title("Department Distribution")
plt.ylabel("")
plt.show()
```

#### Observed:
- Sales contributes 50% of the workforce, indicating it is the dominant department in the organization.

#### Box PLot - Salary Distribution 

```python
plt.boxplot(df['Salary'])
plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show() 
```

#### observed:
- Salaries show moderate spread, with HR having the lowest salary and Marketing the highest, and no extreme outliers present.

# Q : 5 Population vs Sample Statistic 

```python
import numpy as np 
population = np.random.randint(40,100,1000)
sample = np.random.choice(population,100)
```

```python
pop_mean = population.mean()
pop_std = population.std()

sample_mean = sample.mean()
sample_std = sample.std()

print("Population Mean   : ",round(pop_mean,2))
print("Population Sample : ",round(pop_std,2))
print("Sample Mean : ",round(sample_mean,2))
print("Sample Population : ",round(sample_std),2)
```

## Population vs Sample Statistics

- **Population statistics** describe the entire dataset.
- **Sample statistics** describe only a part of the population.


| Aspect | Population Statistics | Sample Statistics |
|------|----------------------|------------------|
| Data Coverage | Entire population | Subset of population |
| Symbols | μ, σ | x̄, s |
| Purpose | Describe whole group | Estimate population |
| Cost & Time | High | Low |
| Accuracy | Exact | Approximate |
