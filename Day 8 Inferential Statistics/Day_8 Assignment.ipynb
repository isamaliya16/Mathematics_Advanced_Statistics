# Statistics Concepts

## 1. Difference Between Descriptive and Inferential Statistics

**Descriptive Statistics**
- Used to summarize and organize data.
- Describes what the data shows.
- Examples: Mean, median, mode, charts, graphs.

**Inferential Statistics**
- Used to make predictions or conclusions about a population based on a sample.
- Involves probability and hypothesis testing.
- Examples: Confidence intervals, hypothesis tests, regression analysis.

---

## 2. Definition of Hypothesis Testing

Hypothesis testing is a statistical method used to make decisions or inferences about a population by analyzing sample data. It helps determine whether there is enough evidence to support a specific claim or assumption.

---

## 3. Null and Alternative Hypotheses for a Delivery-Time Claim

**Claim:**  
The average delivery time of an online service is 30 minutes.

**Null Hypothesis (H₀):**  
The average delivery time is equal to 30 minutes.  
H₀: μ = 30

**Alternative Hypothesis (H₁):**  
The average delivery time is not equal to 30 minutes.  
H₁: μ ≠ 30


# Mini Assignment
####  Claim: Average delivery time is 30 minutes.

- Define H₀ and H₁
- Assume sample data
- Interpret result (conceptual)

## Claim
The average delivery time is **30 minutes**.

---

## Null Hypothesis (H₀)
**H₀: μ = 30 minutes**  
(The average delivery time is equal to 30 minutes.)

---

## Alternative Hypothesis (H₁)
**H₁: μ ≠ 30 minutes**  
(The average delivery time is different from 30 minutes.)

---

## Assumed Sample Data
Suppose we randomly select **50 delivery orders** and find:  
- **Sample mean delivery time = 32 minutes**

---

## Interpretation (Conceptual)
- We compare the sample result with the claim using hypothesis testing.
- If the difference between the sample mean (32 minutes) and the claimed mean (30 minutes) is **statistically significant**, we **reject the null hypothesis**.
- If the difference is **not statistically significant**, we **fail to reject the null hypothesis**.

---

## Conclusion
Based on the sample data, we decide whether there is enough evidence to say that the average delivery time is different from **30 minutes**.


```python
import numpy as np
from scipy import stats

claimed_mean = 30

np.random.seed(1)
sample_data = np.random.normal(loc=32, scale=5, size=50)

sample_mean = np.mean(sample_data)

# One-sample t-test
t_stat, p_value = stats.ttest_1samp(sample_data, claimed_mean)

# Significance level
alpha = 0.05

print("Claimed Mean Delivery Time:", claimed_mean)
print("Sample Mean Delivery Time:", round(sample_mean, 2))
print("p-value:", round(p_value, 4))

if p_value < alpha:
    print("Decision: Reject the null hypothesis.")
    print("Conclusion: Average delivery time is significantly different from 30 minutes.")
else:
    print("Decision: Fail to reject the null hypothesis.")
    print("Conclusion: No significant evidence that delivery time differs from 30 minutes.")
```
