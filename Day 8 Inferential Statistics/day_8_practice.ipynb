## 1️⃣ Types of Statistics

### Descriptive Statistics
- Summarizes data
- Mean, median, SD, charts

### Inferential Statistics
- Makes predictions or decisions
- Uses samples & probability


```python
import numpy as np
import pandas as pd

# Example numeric data
values = np.array([12, 15, 18, 20, 22, 25, 25, 30, 35, 40])
values
```

**Descriptive Statistics mainly includes the followings:**

    "Mean"
    "Median"
    "Mode"
    "Range"
    "Variance (Population)"
    "Std Dev (Population)"
    "Variance (Sample)"
    "Std Dev (Sample)"
    "Quartiles" Q1, Q2, Q3, IQR
    "Skewness"
    "Kurtosis"

Descriptive Statistics

**MEASURES OF CENTRAL TENDENCY**

```python
mean = values.sum() / len(values)
mean
```

```python
sorted_values = np.sort(values)
n = len(sorted_values)

if n % 2 == 0:
    median = (sorted_values[n//2 - 1] + sorted_values[n//2]) / 2
else:
    median = sorted_values[n//2]

median
```

```python
s = pd.Series(values)
mode = s.mode()[0]
mode
```

Descriptive Statistics

**MEASURES OF DISPERSION**

```python
range = values.max() - values.min()
range
```

**Population Variance**
image.png

**Sample Variance**
image.png

```python
variance = ((values - mean) ** 2).sum() / len(values)
print(f"Population Variance: {variance}")

sample_variance = ((values - mean) ** 2).sum() / (len(values) - 1)
print(f"Sample Variance: {sample_variance}")
```

```python
std_dev = variance ** 0.5
print(f"Population Std: {std_dev}")


sample_std = sample_variance ** 0.5
print(f"Sample Std: {sample_std}")
```

Descriptive Statistics

**POSITIONAL STATISTICS**

```python
# Quartiles
Q1, Q2, Q3 = np.percentile(values, [25, 50, 75], interpolation='midpoint')
Q1, Q2, Q3
```

```python
# Interquartile Range (IQR)
IQR = Q3 - Q1
IQR
```

Descriptive Statistics

**SHAPE OF DATA**

image.png

```python
skewness = (3 * (mean - median)) / std_dev
skewness
```

image.png

```python
kurtosis = ((values - mean) ** 4).sum() / (len(values) * (std_dev ** 4))
kurtosis
```

```python
# Final Summary
{
    "Mean": mean,
    "Median": median,
    "Mode": mode,
    "Range": range,
    "Variance (Population)": variance,
    "Std Dev (Population)": std_dev,
    "Variance (Sample)": sample_variance,
    "Std Dev (Sample)": sample_std,
    "Q1": Q1,
    "Q2 (Median)": Q2,
    "Q3": Q3,
    "IQR": IQR,
    "Skewness": skewness,
    "Kurtosis": kurtosis
}
```


## 2️⃣ What Is Inferential Statistics?

Inferential statistics helps us:
- Estimate population parameters
- Test assumptions
- Make decisions with uncertainty


**Inferential Statistics mainly includes the followings:**

    "Sample Mean"
    "Sample Std Dev"
    "Standard Error (SE)"
    "Confidence Interval (CI)"
    "Hypothesis Testing"
    "Z Score"
    "Z Test Decision"
    "T Score"
    "T Test Decision"

```python
import numpy as np

# Sample data (e.g., marks, salaries, heights)
data = np.array([52, 55, 58, 60, 62, 65, 67, 68, 70, 72])

n = len(data)
n
```

```python
# Sample mean
sample_mean = data.sum() / n

# Sample variance
sample_variance = ((data - sample_mean) ** 2).sum() / (n - 1)

# Sample standard deviation
sample_std = sample_variance ** 0.5

sample_mean, sample_variance, sample_std
```


## 3️⃣ What Is Hypothesis Testing?

Hypothesis testing is a structured process to test claims  
about population parameters using sample data.



## 4️⃣ Null Hypothesis (H₀)

Represents:
- No change
- No effect

Example:
H₀: μ = 50,000



## 5️⃣ Alternative Hypothesis (H₁)

Represents:
- Change
- Effect exists

Example:
H₁: μ > 50,000



### Types of Alternative Hypotheses
- Two-tailed
- Right-tailed
- Left-tailed


```python
import numpy as np

# Sample marks of students
marks = np.array([55, 58, 60, 62, 65, 59, 61, 63, 57, 60])
marks
```

```python
sample_mean = marks.sum() / len(marks)
sample_mean
```

```python
# Null Hypothesis (H0)
H0 = "Average marks = 60"

# Alternative Hypothesis (H1)
H1 = "Average marks is NOT 60"

H0, H1
```

```python
# ⚠️ This is NOT a statistical test
# It is only for conceptual understanding

if sample_mean == 60:
    decision = "Data supports Null Hypothesis (H0)"
else:
    decision = "Data does NOT support Null Hypothesis (H0)"

decision
```
