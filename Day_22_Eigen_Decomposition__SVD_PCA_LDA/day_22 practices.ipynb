# 🔹 Part 1: Singular Value Decomposition (SVD)

SVD decomposes a matrix A as:

A = U Σ Vᵀ

Where:
- U → Left singular vectors
- Σ → Singular values
- Vᵀ → Right singular vectors

📘 Applications:
- Image compression
- Recommendation systems
- Noise reduction


image.png

```python
# Singular Value Decomposition (SVD)

import numpy as np

# Define matrix
A = np.array([
    [3, 1],
    [1, 3]
])

# -----------------------------
# SVD Decomposition
# -----------------------------
U, S, VT = np.linalg.svd(A)

# Convert singular values to diagonal matrix
Sigma = np.zeros_like(A, dtype=float)
np.fill_diagonal(Sigma, S)

# Reconstruct matrix
A_reconstructed = U @ Sigma @ VT

# -----------------------------
# Output
# -----------------------------
print("Matrix A:\n", A)

print("\nMatrix U:\n", U)

print("\nSingular Values:", S)

print("\nMatrix V^T:\n", VT)

print("\nReconstructed Matrix:\n", A_reconstructed)
```

# 🔹 Part 2: Principal Component Analysis (PCA)

Principal Component Analysis (PCA) is a technique used to:
- Reduce the number of features (dimensions)
- Keep maximum information (variance) in the data.

It transforms original variables into new uncorrelated variables called Principal Components.


image.png

```python
# Principal Component Analysis (PCA)

import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# -----------------------------
# Dataset
# -----------------------------
X = np.array([
    [170, 65],
    [165, 60],
    [180, 75],
    [175, 70]
])

# -----------------------------
# Step 1: Standardize Data
# -----------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# -----------------------------
# Step 2: Apply PCA
# -----------------------------
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# -----------------------------
# Output
# -----------------------------
print("Original Data:\n", X)

print("\nStandardized Data:\n", X_scaled)

print("\nPrincipal Components:\n", pca.components_)

print("\nExplained Variance Ratio:\n", pca.explained_variance_ratio_)

print("\nTransformed Data:\n", X_pca)
```

# 🔹 Part 3: Linear Discriminant Analysis (LDA)

Linear Discriminant Analysis (LDA) is a technique used for:
- Dimensionality Reduction
- Classification

It finds directions that maximize separation between classes.

```python
# Linear Discriminant Analysis (LDA)

import numpy as np
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# -----------------------------
# Dataset
# -----------------------------
X = np.array([
    [170, 65],
    [175, 70],
    [160, 55],
    [165, 60]
])

y = np.array([0, 0, 1, 1])  # class labels

# -----------------------------
# Apply LDA
# -----------------------------
lda = LinearDiscriminantAnalysis(n_components=1)

X_lda = lda.fit_transform(X, y)

# -----------------------------
# Output
# -----------------------------
print("Original Data:\n", X)

print("\nClass Labels:\n", y)

print("\nLDA Projection:\n", X_lda)

print("\nLDA Coefficients:\n", lda.coef_)
```

# 🔹 PCA vs LDA

|------|------|------|
| Type | Unsupervised | Supervised |
| Goal | Maximize variance | Maximize class separation |
| Uses labels | No | Yes |
