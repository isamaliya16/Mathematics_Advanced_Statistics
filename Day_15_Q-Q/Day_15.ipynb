# Q : 1
### Normal Distribution Heights
#### A dataset of 500 students' heights is approximately normal with
- Mean = 165 cm,
- Standard deviation = 10 cm.

##### Find the probability a randomly chosen student has height:
- between 160-170 cm
- greater than 180 cm
##### Sketch the curve with shaded regions.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm 

mean = 165
std_dev = 10 
x = np.linspace(120,210,1000)
pdf = norm.pdf(x, mean, std_dev)

prob_between = norm.cdf(170,mean, std_dev) - norm.cdf(160, mean, std_dev)
prob_greater = 1 - norm.cdf(180,  mean, std_dev)

print("Probability (160 cm ≤ height ≤ 170 cm):", round(prob_between, 4))
print("Probability (height > 180 cm):", round(prob_greater, 4))

plt.figure()
plt.plot(x, pdf)

mask_between = (x >= 160) & (x <= 170)
plt.fill_between(x[mask_between], pdf[mask_between],0)

mask_greater = x >= 180
plt.fill_between(x[mask_greater],pdf[mask_greater],0)

plt.title("Normal Distribution of Heights ")
plt.xlabel("Height (cm)")
plt.ylabel("Density")
plt.show()
```

# Q.2
## Exponential Distribution - Service Times
### Customers arrive at a service desk. Service times follow an exponential distribution with mean service time = 5 minutes.
#### Compute probability service time is:
- less than 3 minutes
- more than 10 minutes
#### Plot the probability density curve.



```python
from scipy.stats import expon

mean_time = 5 
scale = mean_time
prob_less_3 =  expon.cdf(3,scale=scale)
prob_more_10 = 1 - expon.cdf(10, scale=scale)

print("Probability (service time < 3 minutes):", round(prob_less_3, 4))
print("Probability (service time > 10 minutes):", round(prob_more_10, 4))

x = np.linspace(0, 25, 1000)
pdf = expon.pdf(x, scale = scale)

plt.figure()
plt.plot(x, pdf)

mask_less = x <= 3
plt.fill_between(x[mask_less], pdf[mask_less], 0)

mask_more = x >= 10
plt.fill_between(x[mask_more], pdf[mask_more], 0)

plt.title("Exponential Distribution - Service Times")
plt.xlabel("Service Time (minutes)")
plt.ylabel("Density")
plt.show()
```

# Q.3
- Normality Check with Exam Scores
- Exam scores of 100 students are given.
- Construct a Q-Q plot against a normal distribution.
- Comment if the dataset is approximately normal.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
scores = np.random.normal(loc=70, scale=10, size=100)

plt.figure()
stats.probplot(scores, dist="norm", plot=plt)

plt.title("Q-Q Plot: Exam Scores vs Normal Distribution")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")
plt.show()

corr = stats.probplot(scores, dist="norm")[1][2]
if corr > 0.97:
    print("Dataset appears approximately normal.")
else:
    print("Dataset deviates from normality.")
```

# Q.4
1. Stock Returns Q-Q Plot
2. Daily returns (%) of a stock for 250 trading days are collected.
3. Generate a Q-Q plot comparing data with normal distribution.
4. Identify presence of heavy tails or skewness.

```python
np.random.seed(7)
returns = stats.t.rvs(df=3, size=250) * 2

plt.figure()
stats.probplot(returns, dist="norm", plot=plt)

plt.title("Q-Q Plot: Stock Returns vs Normal Distribution")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")
plt.show()

skewness = stats.skew(returns)
kurtosis = stats.kurtosis(returns)

print("Skewness:", round(skewness, 3))
print("Excess Kurtosis:", round(kurtosis, 3))

if abs(kurtosis) > 1:
    print("Heavy tails detected.")
else:
    print("No strong heavy tails detected.")

if abs(skewness) > 0.5:
    print("Noticeable skewness present.")
else:
    print("Distribution is roughly symmetric.")
```

## Q.5 — Classify & Compare: Discrete vs Continuous

### Classification with Explanation

| Scenario | Type | Explanation |
|----------|------|-------------|
| Number of goals in a football match | **Discrete** | Goals are counted in whole numbers (0, 1, 2…). Fractions are not possible. |
| Weight of newborn babies | **Continuous** | Weight is measured and can take any decimal value within a range. |
| Dice outcomes | **Discrete** | Dice results are fixed integers (1–6). No fractional values exist. |
| Lifespan of light bulbs | **Continuous** | Time can be measured precisely and may take any real value. |

### Key Concept

- **Discrete distribution → counting values**
- **Continuous distribution → measured values**

Discrete data consists of separate, countable outcomes, while continuous data can take infinitely many values within a range.

## Q.7
1. Coin Toss (Bernoulli)
2. Toss a coin once.
3. Define success = "Head"
4. Find probability mass function (PMF).
5. Compute expected value and variance
6. Plot the Bernoulli distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.5 
x = np.array([0, 1])  # 0 = Tail, 1 = Head

pmf = np.array([1 - p, p])

expected_value = p
variance = p * (1 - p)

print("PMF:")
print(f"P(X=0) = {pmf[0]}")
print(f"P(X=1) = {pmf[1]}")
print("\nExpected Value:", expected_value)
print("Variance:", variance)
```

```python

plt.figure()
plt.bar(x, pmf)
plt.title("Bernoulli Distribution (Coin Toss)")
plt.xlabel("Outcome (0=Tail, 1=Head)")
plt.ylabel("Probability")
plt.show()
```

## Q.8

1. Exam Pass Probability (Binomial)
2. Each student has a 0.6 probability of passing an exam. In a group of 10 students:
3. Find probability exactly 7 pass.
4. Find probability at least 8 pass.
5. Plot the Binomial distribution.

```python
from scipy.stats import binom 
n = 10 
p = 0.6
prob_exact_7 = binom.pmf(7, n, p)
prob_at_least_8 = binom.pmf(8, n, p) + binom.pmf(9, n, p) + binom.pmf(10, n, p)

print("p(exactly 7 pass) : ",round(prob_exact_7,4))
print("p(at least 8 pass) : ",round(prob_at_least_8,4))

x = np.arange(0, n + 1)
pmf = binom.pmf(x, n, p)
```

```python
plt.figure()
plt.bar(x, pmf) 
plt.title("Binomial Distribution (n=10, p=0.6)")
plt.xlabel("Number of Students Passing")
plt.ylabel("Probability")
plt.show()
```
