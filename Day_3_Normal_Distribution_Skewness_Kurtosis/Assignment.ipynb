## 1️⃣ What is Gaussian Distribution?

Gaussian distribution is a continuous probability distribution that is symmetric and bell-shaped.

It is also called the **Normal Distribution**.

Main properties:
- Mean = Median = Mode
- Symmetrical around the mean
- Defined by Mean (μ) and Standard Deviation (σ)
- Total area under the curve = 1

Example:
Heights, IQ scores, exam marks often follow normal distribution.

---

## 2️⃣ Define Skewness

Skewness measures the asymmetry of a data distribution.

Types:
- **Zero Skewness** → Symmetrical distribution
- **Positive Skewness** → Tail on right side (Mean > Median)
- **Negative Skewness** → Tail on left side (Mean < Median)

👉 It tells whether data is tilted left or right.

---

## 3️⃣ Difference Between Skewness and Kurtosis

| Basis | Skewness | Kurtosis |
|--------|-----------|-----------|
| Meaning | Measures asymmetry | Measures peakedness or tail heaviness |
| Focus | Direction of data | Shape of peak and tails |
| Indicates | Left or right tilt | How sharp or flat the distribution is |
| Example | Income data often positively skewed | Extreme outliers affect kurtosis |

---

## 4️⃣ Why Kurtosis is Important?

Kurtosis is important because:

- It shows presence of outliers
- Helps understand tail behavior
- Useful in financial risk analysis
- Identifies extreme values in dataset

Types:
- **Mesokurtic** → Normal shape
- **Leptokurtic** → High peak, heavy tails (more outliers)
- **Platykurtic** → Flat peak, light tails

👉 Kurtosis helps in understanding risk and data stability.

# Assignement 
## Generate salary data of 1000 employees.
## Tasks:
- Check normality visually.
- Compute skewness and kurtosis.
- Interpret results.

```python
import numpy as np
import matplotlib.pyplot as plt 
from scipy.stats import skew,kurtosis

np.random.seed(42)
salary = np.random.lognormal(mean=10, sigma=0.4, size=1000)
```

```python
plt.figure(figsize=(10, 6))
plt.hist(salary, bins=30, density=True, alpha=0.7)
plt.axvline(np.mean(salary), linestyle='--', linewidth=2)
plt.title("Salary Distribution (1000 Employees)", fontsize=14, fontweight="bold")
plt.xlabel("Salary")
plt.ylabel("Density")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
salary_sk = skew(salary)
salary_kurtosis = kurtosis(salary,fisher=False)
print("Skewness : ", round(salary_sk,4))
print("Kurtosis : ", round(salary_kurtosis,4))
```

## Interpretation 🧠 (Exam / Interview Ready)
#### 🔹 Skewness

Skewness > 0 → positively skewed

Indicates most employees earn moderate salaries,
but few employees earn very high salaries

#### 🔹 Kurtosis

- Kurtosis > 3 → Leptokurtic
- Distribution has:
- Sharp peak
- Heavy tails
- Extreme salary values (very high salaries) exist

#### ✅ Final Conclusion
The salary data of 1000 employees is positively skewed and leptokurtic, indicating income inequality where a small number of employees earn significantly higher salaries than the majority.
